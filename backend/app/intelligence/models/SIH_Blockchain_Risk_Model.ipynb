```python
import pandas as pd

features = pd.read_csv(
    "EllipticPlusPlus/Actors Dataset/wallets_features.csv"
)

classes = pd.read_csv(
    "EllipticPlusPlus/Actors Dataset/wallets_classes.csv"
)

print("Features:", features.shape)
print("Classes:", classes.shape)

display(features.head())
display(classes.head())
```

```python
# ============================================
# STEP 1: Inspect Elliptic++ dataset
# ============================================

print("FEATURE SHAPE:", features.shape)
print("CLASS SHAPE:", classes.shape)

print("\nFEATURE COLUMNS:")
for i, col in enumerate(features.columns):
    print(i, "→", col)

print("\nCLASS COLUMNS:")
for i, col in enumerate(classes.columns):
    print(i, "→", col)

print("\nCLASS DISTRIBUTION:")
print(classes["class"].value_counts(dropna=False))

print("\nFEATURE DATA TYPES:")
print(features.dtypes)

print("\nMISSING VALUES:")
print(features.isna().sum().sort_values(ascending=False).head(20))
```

```python
# ============================================
# STEP 2: Check address/label relationship
# ============================================

feature_addresses = set(features["address"])
class_addresses = set(classes["address"])

print("Unique feature addresses:", len(feature_addresses))
print("Unique class addresses:", len(class_addresses))

print(
    "Feature addresses with labels:",
    len(feature_addresses & class_addresses)
)

print(
    "Feature addresses without labels:",
    len(feature_addresses - class_addresses)
)
```

```python
# ============================================================
# STEP 3 — Prepare Elliptic++ for ML
# ============================================================

import pandas as pd
import numpy as np

# Make copies so we don't accidentally modify the originals
X_df = features.copy()
y_df = classes.copy()

print("Original feature rows:", len(X_df))
print("Original wallets:", y_df["address"].nunique())

# ------------------------------------------------------------
# 1. Join wallet labels onto temporal feature records
# ------------------------------------------------------------

df = X_df.merge(
    y_df[["address", "class"]],
    on="address",
    how="inner"
)

print("Joined dataset:", df.shape)

# ------------------------------------------------------------
# 2. Keep only labelled wallets
#
# Elliptic++:
# 1 = illicit
# 2 = licit
# 3 = unknown
# ------------------------------------------------------------

df = df[df["class"].isin([1, 2])].copy()

# Convert:
# illicit = 1
# licit   = 0

df["target"] = (df["class"] == 1).astype(int)

print("\nAfter removing unknown wallets:")
print(df.shape)

print("\nClass distribution:")
print(df["target"].value_counts())

print("\nClass percentages:")
print(df["target"].value_counts(normalize=True) * 100)

# ------------------------------------------------------------
# 3. Separate address/labels from ML features
# ------------------------------------------------------------

X = df.drop(columns=["address", "class", "target"])
y = df["target"]

# Keep only numeric columns
X = X.select_dtypes(include=[np.number])

# Replace problematic values
X = X.replace([np.inf, -np.inf], np.nan)
X = X.fillna(0)

print("\nFinal ML matrix:")
print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nFeatures being used:")
print(X.columns.tolist())
```

```python
# ============================================================
# STEP 4 — Wallet-level train/test split
# ============================================================

from sklearn.model_selection import train_test_split

# Get unique labelled wallets and their labels
wallet_labels = df[["address", "target"]].drop_duplicates("address")

train_wallets, test_wallets = train_test_split(
    wallet_labels,
    test_size=0.20,
    stratify=wallet_labels["target"],
    random_state=42
)

train_addresses = set(train_wallets["address"])
test_addresses = set(test_wallets["address"])

# Get corresponding rows from the temporal feature dataset
train_mask = df["address"].isin(train_addresses)
test_mask = df["address"].isin(test_addresses)

X_train = X.loc[train_mask]
X_test = X.loc[test_mask]

y_train = y.loc[train_mask]
y_test = y.loc[test_mask]

print("TRAIN:")
print("X:", X_train.shape)
print("y:", y_train.shape)

print("\nTEST:")
print("X:", X_test.shape)
print("y:", y_test.shape)

print("\nTrain class distribution:")
print(y_train.value_counts())

print("\nTest class distribution:")
print(y_test.value_counts())
```

```python
# ============================================================
# STEP 5 — Train XGBoost Blockchain Risk Model
# ============================================================

from xgboost import XGBClassifier

# Calculate class imbalance
negative = (y_train == 0).sum()
positive = (y_train == 1).sum()

scale_pos_weight = negative / positive

print("Legitimate:", negative)
print("Illicit:", positive)
print("Scale pos weight:", scale_pos_weight)

model = XGBClassifier(
    n_estimators=250,
    max_depth=6,
    learning_rate=0.08,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="binary:logistic",
    eval_metric="aucpr",
    scale_pos_weight=scale_pos_weight,
    random_state=42,
    n_jobs=-1,
    tree_method="hist"
)

print("\nTraining model...")

model.fit(
    X_train,
    y_train,
    eval_set=[(X_test, y_test)],
    verbose=True
)

print("\nTRAINING COMPLETE.")
```

```python
# ============================================================
# STEP 6 — Evaluate
# ============================================================

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score,
    average_precision_score
)

# Predictions
y_pred = model.predict(X_test)

# Probability of being illicit
y_prob = model.predict_proba(X_test)[:, 1]

print("=" * 60)
print("CLASSIFICATION REPORT")
print("=" * 60)

print(
    classification_report(
        y_test,
        y_pred,
        target_names=["LEGITIMATE", "ILLICIT"],
        digits=4
    )
)

print("=" * 60)
print("CONFUSION MATRIX")
print("=" * 60)

print(confusion_matrix(y_test, y_pred))

print("=" * 60)
print("METRICS")
print("=" * 60)

print("ROC-AUC :", round(roc_auc_score(y_test, y_prob), 4))
print("PR-AUC  :", round(average_precision_score(y_test, y_prob), 4))
```

```python
# ============================================================
# STEP 7 — Feature Importance
# ============================================================

importance = pd.Series(
    model.feature_importances_,
    index=X.columns
).sort_values(ascending=False)

print("TOP 20 FEATURES:")
display(importance.head(20))
```

```python
# ============================================================
# MODEL 1 — OFFICIAL ELLIPTIC++ STYLE PREPARATION
# ============================================================

import pandas as pd
import numpy as np

# Use the combined wallet feature/class dataset if available.
# Otherwise create it from our existing data.

combined = features.merge(
    classes,
    on="address",
    how="inner"
)

print("Combined:", combined.shape)

# ------------------------------------------------------------
# Keep only known labels
# 2 = LICIT
# 1 = ILLICIT
# ------------------------------------------------------------

combined = combined[
    combined["class"].isin([1, 2])
].copy()

# Official binary convention:
# class 2 = 0 (licit)
# class 1 = 1 (illicit)

combined["target"] = (
    combined["class"] == 1
).astype(int)

print("\nClass distribution:")
print(combined["target"].value_counts())

print("\nPercentages:")
print(
    (combined["target"].value_counts(normalize=True) * 100)
    .round(2)
)
```

```python
# ============================================================
# WALLET-LEVEL FEATURE AGGREGATION
# ============================================================

# Separate numeric features
numeric_cols = combined.select_dtypes(
    include=[np.number]
).columns.tolist()

# Don't aggregate labels
numeric_cols = [
    c for c in numeric_cols
    if c not in ["class", "target"]
]

print("Numeric features:", len(numeric_cols))

# Aggregate temporal observations for each wallet
wallet_features = (
    combined
    .groupby("address")[numeric_cols]
    .agg([
        "mean",
        "max",
        "min"
    ])
)

# Flatten multi-level column names
wallet_features.columns = [
    f"{col}_{stat}"
    for col, stat in wallet_features.columns
]

wallet_features = wallet_features.reset_index()

# Get one label per wallet
wallet_labels = (
    combined[["address", "target"]]
    .drop_duplicates("address")
)

# Join labels
wallet_data = wallet_features.merge(
    wallet_labels,
    on="address",
    how="inner"
)

print("\nWallet-level dataset:")
print(wallet_data.shape)

print("\nClass distribution:")
print(wallet_data["target"].value_counts())
```

```python
# ============================================================
# MODEL 1 — WALLET BEHAVIOR XGBOOST
# ============================================================

from sklearn.model_selection import train_test_split
from xgboost import XGBClassifier

# Separate X/y
X_wallet = wallet_data.drop(
    columns=["address", "target"]
)

y_wallet = wallet_data["target"]

# Clean
X_wallet = X_wallet.replace(
    [np.inf, -np.inf],
    np.nan
)

X_wallet = X_wallet.fillna(0)

# ------------------------------------------------------------
# Split at wallet level
# ------------------------------------------------------------

X_train2, X_test2, y_train2, y_test2 = train_test_split(
    X_wallet,
    y_wallet,
    test_size=0.30,
    stratify=y_wallet,
    random_state=42
)

print("Training:", X_train2.shape)
print("Testing :", X_test2.shape)

print("\nTrain labels:")
print(y_train2.value_counts())

# ------------------------------------------------------------
# Class weighting
# ------------------------------------------------------------

negative = (y_train2 == 0).sum()
positive = (y_train2 == 1).sum()

weight = negative / positive

print("\nClass weight:", weight)

# ------------------------------------------------------------
# Model
# ------------------------------------------------------------

model2 = XGBClassifier(
    n_estimators=500,
    max_depth=5,
    learning_rate=0.05,

    subsample=0.85,
    colsample_bytree=0.85,

    objective="binary:logistic",
    eval_metric="aucpr",

    scale_pos_weight=weight,

    tree_method="hist",
    n_jobs=-1,
    random_state=42
)

print("\nTraining Model 1...")

model2.fit(
    X_train2,
    y_train2,
    eval_set=[(X_test2, y_test2)],
    verbose=False
)

print("\nMODEL 1 COMPLETE")
```

```python
# ============================================================
# MODEL 1 — EVALUATION
# ============================================================

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score,
    average_precision_score
)

pred2 = model2.predict(X_test2)

prob2 = model2.predict_proba(
    X_test2
)[:, 1]

print("=" * 70)
print("MODEL 1 — WALLET BEHAVIOR XGBOOST")
print("=" * 70)

print(
    classification_report(
        y_test2,
        pred2,
        target_names=[
            "LEGITIMATE",
            "ILLICIT"
        ],
        digits=4
    )
)

print("\nConfusion Matrix:")
print(confusion_matrix(y_test2, pred2))

print("\nROC-AUC:")
print(round(
    roc_auc_score(y_test2, prob2),
    4
))

print("\nPR-AUC:")
print(round(
    average_precision_score(y_test2, prob2),
    4
))
```

```python
# ==========================================
# MODEL 1 — RISK THRESHOLD ANALYSIS
# ==========================================

from sklearn.metrics import precision_score, recall_score, f1_score

thresholds = [0.30, 0.40, 0.50, 0.60, 0.70, 0.80, 0.90]

print("Threshold | Precision | Recall | F1")
print("-" * 42)

for threshold in thresholds:
    pred_threshold = (prob2 >= threshold).astype(int)

    precision = precision_score(y_test2, pred_threshold)
    recall = recall_score(y_test2, pred_threshold)
    f1 = f1_score(y_test2, pred_threshold)

    print(
        f"{threshold:9.2f} | "
        f"{precision:9.4f} | "
        f"{recall:6.4f} | "
        f"{f1:6.4f}"
    )
```

```python
# ============================================================
# MODEL 1.1 — TEMPORAL VALIDATION
# ============================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. Start from ORIGINAL temporal data
# ------------------------------------------------------------

temporal_data = features.merge(
    classes[["address", "class"]],
    on="address",
    how="inner"
)

# Only known wallets
temporal_data = temporal_data[
    temporal_data["class"].isin([1, 2])
].copy()

# Binary label
temporal_data["target"] = (
    temporal_data["class"] == 1
).astype(int)

# ------------------------------------------------------------
# 2. Find the FIRST time step each wallet appears
# ------------------------------------------------------------

wallet_first_time = (
    temporal_data
    .groupby("address")["Time step"]
    .min()
    .reset_index(name="first_time_step")
)

wallet_first_time = wallet_first_time.merge(
    temporal_data[["address", "target"]]
    .drop_duplicates("address"),
    on="address",
    how="left"
)

print("Total labelled wallets:", len(wallet_first_time))

# ------------------------------------------------------------
# 3. Automatically choose a chronological cutoff
#    around 70% of wallets for training
# ------------------------------------------------------------

cutoff = int(
    wallet_first_time["first_time_step"]
    .quantile(0.70)
)

print("Temporal cutoff:", cutoff)

# Early wallets = training
train_wallets = wallet_first_time[
    wallet_first_time["first_time_step"] <= cutoff
].copy()

# Later wallets = testing
test_wallets = wallet_first_time[
    wallet_first_time["first_time_step"] > cutoff
].copy()

train_addresses = set(train_wallets["address"])
test_addresses = set(test_wallets["address"])

print("\nTRAIN WALLETS:", len(train_addresses))
print("TEST WALLETS :", len(test_addresses))

print("\nTRAIN LABELS:")
print(train_wallets["target"].value_counts())

print("\nTEST LABELS:")
print(test_wallets["target"].value_counts())

# ------------------------------------------------------------
# 4. IMPORTANT:
# Train features only use observations available
# up to the temporal cutoff.
#
# Test wallets are completely unseen during training.
# ------------------------------------------------------------

train_rows = temporal_data[
    (temporal_data["address"].isin(train_addresses)) &
    (temporal_data["Time step"] <= cutoff)
].copy()

test_rows = temporal_data[
    temporal_data["address"].isin(test_addresses)
].copy()

print("\nTRAIN TEMPORAL ROWS:", train_rows.shape)
print("TEST TEMPORAL ROWS :", test_rows.shape)
```

```python
# ============================================================
# MODEL 1.1 — TEMPORAL FEATURE AGGREGATION
# ============================================================

# Numeric columns from the original dataset
numeric_cols = temporal_data.select_dtypes(
    include=[np.number]
).columns.tolist()

# Don't aggregate labels
numeric_cols = [
    c for c in numeric_cols
    if c not in ["class", "target"]
]

print("Original numeric features:", len(numeric_cols))

# ------------------------------------------------------------
# TRAIN AGGREGATION
# ------------------------------------------------------------

train_features = (
    train_rows
    .groupby("address")[numeric_cols]
    .agg(["mean", "max", "min"])
)

train_features.columns = [
    f"{col}_{stat}"
    for col, stat in train_features.columns
]

train_features = train_features.reset_index()

# ------------------------------------------------------------
# TEST AGGREGATION
# ------------------------------------------------------------

test_features = (
    test_rows
    .groupby("address")[numeric_cols]
    .agg(["mean", "max", "min"])
)

test_features.columns = [
    f"{col}_{stat}"
    for col, stat in test_features.columns
]

test_features = test_features.reset_index()

# ------------------------------------------------------------
# ADD LABELS
# ------------------------------------------------------------

train_data = train_features.merge(
    train_wallets[["address", "target"]],
    on="address",
    how="inner"
)

test_data = test_features.merge(
    test_wallets[["address", "target"]],
    on="address",
    how="inner"
)

print("\nTRAIN DATA:", train_data.shape)
print("TEST DATA :", test_data.shape)

print("\nTrain labels:")
print(train_data["target"].value_counts())

print("\nTest labels:")
print(test_data["target"].value_counts())
```

```python
# ============================================================
# MODEL 1.1 — TEMPORAL XGBOOST
# ============================================================

from xgboost import XGBClassifier

# Remove identifiers and labels
X_temporal_train = train_data.drop(
    columns=["address", "target"]
)

y_temporal_train = train_data["target"]

X_temporal_test = test_data.drop(
    columns=["address", "target"]
)

y_temporal_test = test_data["target"]

# Clean
X_temporal_train = (
    X_temporal_train
    .replace([np.inf, -np.inf], np.nan)
    .fillna(0)
)

X_temporal_test = (
    X_temporal_test
    .replace([np.inf, -np.inf], np.nan)
    .fillna(0)
)

# Make absolutely sure feature order is identical
X_temporal_test = X_temporal_test[
    X_temporal_train.columns
]

# ------------------------------------------------------------
# Class imbalance
# ------------------------------------------------------------

negative = int(
    (y_temporal_train == 0).sum()
)

positive = int(
    (y_temporal_train == 1).sum()
)

temporal_weight = negative / positive

print("Legitimate:", negative)
print("Illicit   :", positive)
print("Weight    :", temporal_weight)

# ------------------------------------------------------------
# MODEL
# ------------------------------------------------------------

temporal_model = XGBClassifier(
    n_estimators=500,
    max_depth=5,
    learning_rate=0.05,

    subsample=0.85,
    colsample_bytree=0.85,

    objective="binary:logistic",
    eval_metric="aucpr",

    scale_pos_weight=temporal_weight,

    tree_method="hist",
    n_jobs=-1,
    random_state=42
)

print("\nTraining temporal model...")

temporal_model.fit(
    X_temporal_train,
    y_temporal_train,
    eval_set=[
        (X_temporal_test, y_temporal_test)
    ],
    verbose=False
)

print("\nTEMPORAL MODEL COMPLETE")
```

```python
# ============================================================
# MODEL 1.1 — TEMPORAL EVALUATION
# ============================================================

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score,
    average_precision_score
)

temporal_pred = temporal_model.predict(
    X_temporal_test
)

temporal_prob = temporal_model.predict_proba(
    X_temporal_test
)[:, 1]

print("=" * 70)
print("TEMPORAL MODEL RESULTS")
print("=" * 70)

print("\nCLASSIFICATION REPORT")
print("-" * 70)

print(
    classification_report(
        y_temporal_test,
        temporal_pred,
        target_names=[
            "LEGITIMATE",
            "ILLICIT"
        ],
        digits=4
    )
)

print("\nCONFUSION MATRIX")
print("-" * 70)

print(
    confusion_matrix(
        y_temporal_test,
        temporal_pred
    )
)

print("\nMETRICS")
print("-" * 70)

temporal_roc = roc_auc_score(
    y_temporal_test,
    temporal_prob
)

temporal_pr = average_precision_score(
    y_temporal_test,
    temporal_prob
)

print(f"ROC-AUC : {temporal_roc:.4f}")
print(f"PR-AUC  : {temporal_pr:.4f}")

print("\n" + "=" * 70)
```

```python
# ============================================================
# MODEL 2 — FORWARD TEMPORAL VALIDATION
# ============================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. Original temporal dataset + labels
# ------------------------------------------------------------

temporal_data = features.merge(
    classes[["address", "class"]],
    on="address",
    how="inner"
)

# Keep only labelled wallets
temporal_data = temporal_data[
    temporal_data["class"].isin([1, 2])
].copy()

# 1 = illicit
# 2 = legitimate
temporal_data["target"] = (
    temporal_data["class"] == 1
).astype(int)

# ------------------------------------------------------------
# 2. Define chronological windows
# ------------------------------------------------------------

TRAIN_END = 29
VAL_START = 30
VAL_END = 34
TEST_START = 35
TEST_END = 49

train_rows = temporal_data[
    temporal_data["Time step"] <= TRAIN_END
].copy()

val_rows = temporal_data[
    (temporal_data["Time step"] >= VAL_START) &
    (temporal_data["Time step"] <= VAL_END)
].copy()

test_rows = temporal_data[
    (temporal_data["Time step"] >= TEST_START) &
    (temporal_data["Time step"] <= TEST_END)
].copy()

print("TRAIN:", train_rows.shape)
print("VALIDATION:", val_rows.shape)
print("TEST:", test_rows.shape)

print("\nTime ranges:")
print("Train:", train_rows["Time step"].min(),
      "→", train_rows["Time step"].max())

print("Validation:", val_rows["Time step"].min(),
      "→", val_rows["Time step"].max())

print("Test:", test_rows["Time step"].min(),
      "→", test_rows["Time step"].max())
```

```python
# ============================================================
# MODEL 2 — TEMPORAL FEATURE ENGINEERING
# ============================================================

# Original numerical feature columns
numeric_cols = temporal_data.select_dtypes(
    include=[np.number]
).columns.tolist()

# Remove labels and time itself
numeric_cols = [
    c for c in numeric_cols
    if c not in ["class", "target", "Time step"]
]

print("Base numeric features:", len(numeric_cols))


def aggregate_window(rows):
    """
    Turn temporal observations into one behavioral vector
    per wallet for a specific time window.
    """

    agg = (
        rows
        .groupby("address")[numeric_cols]
        .agg(["mean", "max", "min"])
    )

    agg.columns = [
        f"{feature}_{stat}"
        for feature, stat in agg.columns
    ]

    return agg.reset_index()


train_features = aggregate_window(train_rows)
val_features = aggregate_window(val_rows)
test_features = aggregate_window(test_rows)


# ------------------------------------------------------------
# Add wallet labels
# ------------------------------------------------------------

wallet_labels = (
    temporal_data[["address", "target"]]
    .drop_duplicates("address")
)

train_data = train_features.merge(
    wallet_labels,
    on="address",
    how="inner"
)

val_data = val_features.merge(
    wallet_labels,
    on="address",
    how="inner"
)

test_data = test_features.merge(
    wallet_labels,
    on="address",
    how="inner"
)


print("\nAggregated datasets:")
print("Train:", train_data.shape)
print("Validation:", val_data.shape)
print("Test:", test_data.shape)

print("\nTrain labels:")
print(train_data["target"].value_counts())

print("\nValidation labels:")
print(val_data["target"].value_counts())

print("\nTest labels:")
print(test_data["target"].value_counts())
```

```python
# ============================================================
# MODEL 2 — TEMPORAL XGBOOST
# ============================================================

from xgboost import XGBClassifier

# ------------------------------------------------------------
# Prepare matrices
# ------------------------------------------------------------

X_train_t = train_data.drop(
    columns=["address", "target"]
)

y_train_t = train_data["target"]

X_val_t = val_data.drop(
    columns=["address", "target"]
)

y_val_t = val_data["target"]

X_test_t = test_data.drop(
    columns=["address", "target"]
)

y_test_t = test_data["target"]


# ------------------------------------------------------------
# Clean numerical values
# ------------------------------------------------------------

for df_ in [X_train_t, X_val_t, X_test_t]:

    df_.replace(
        [np.inf, -np.inf],
        np.nan,
        inplace=True
    )

    df_.fillna(0, inplace=True)


# Make sure every dataset has identical columns
X_val_t = X_val_t.reindex(
    columns=X_train_t.columns,
    fill_value=0
)

X_test_t = X_test_t.reindex(
    columns=X_train_t.columns,
    fill_value=0
)


# ------------------------------------------------------------
# Handle class imbalance
# ------------------------------------------------------------

negative = int((y_train_t == 0).sum())
positive = int((y_train_t == 1).sum())

temporal_weight = negative / positive

print("Legitimate:", negative)
print("Illicit:", positive)
print("Class weight:", temporal_weight)


# ------------------------------------------------------------
# XGBoost
# ------------------------------------------------------------

model_temporal = XGBClassifier(
    n_estimators=500,
    max_depth=5,
    learning_rate=0.05,

    subsample=0.85,
    colsample_bytree=0.85,

    objective="binary:logistic",
    eval_metric="aucpr",

    scale_pos_weight=temporal_weight,

    tree_method="hist",
    n_jobs=-1,
    random_state=42
)


print("\nTraining Model 2...")

model_temporal.fit(
    X_train_t,
    y_train_t,

    eval_set=[
        (X_val_t, y_val_t)
    ],

    verbose=False
)

print("\nMODEL 2 TRAINING COMPLETE")
```

```python
# ============================================================
# MODEL 2 — VALIDATION
# ============================================================

from sklearn.metrics import (
    classification_report,
    roc_auc_score,
    average_precision_score
)

val_prob = model_temporal.predict_proba(
    X_val_t
)[:, 1]

print("=" * 60)
print("VALIDATION")
print("=" * 60)

print(
    classification_report(
        y_val_t,
        (val_prob >= 0.50).astype(int),
        target_names=[
            "LEGITIMATE",
            "ILLICIT"
        ],
        digits=4
    )
)

print(
    "ROC-AUC:",
    round(
        roc_auc_score(y_val_t, val_prob),
        4
    )
)

print(
    "PR-AUC:",
    round(
        average_precision_score(y_val_t, val_prob),
        4
    )
)
```

```python
# ============================================================
# MODEL 2 — THRESHOLD OPTIMIZATION
# ============================================================

from sklearn.metrics import precision_score, recall_score, f1_score

thresholds = np.arange(
    0.10,
    0.96,
    0.05
)

results = []

for threshold in thresholds:

    pred = (
        val_prob >= threshold
    ).astype(int)

    precision = precision_score(
        y_val_t,
        pred,
        zero_division=0
    )

    recall = recall_score(
        y_val_t,
        pred,
        zero_division=0
    )

    f1 = f1_score(
        y_val_t,
        pred,
        zero_division=0
    )

    results.append({
        "threshold": round(float(threshold), 2),
        "precision": precision,
        "recall": recall,
        "f1": f1
    })


threshold_df = pd.DataFrame(results)

print(
    threshold_df
    .sort_values("f1", ascending=False)
    .head(10)
)
```

```python
best_threshold = float(
    threshold_df.loc[
        threshold_df["f1"].idxmax(),
        "threshold"
    ]
)

print(
    "\nSelected threshold:",
    best_threshold
)
```

```python
# ============================================================
# MODEL 2 — FINAL UNSEEN FUTURE PERIOD
# ============================================================

test_prob = model_temporal.predict_proba(
    X_test_t
)[:, 1]

test_pred = (
    test_prob >= best_threshold
).astype(int)

print("=" * 70)
print("MODEL 2 — FINAL TEMPORAL TEST")
print("=" * 70)

print(
    classification_report(
        y_test_t,
        test_pred,
        target_names=[
            "LEGITIMATE",
            "ILLICIT"
        ],
        digits=4
    )
)

print("\nROC-AUC:",
      round(
          roc_auc_score(y_test_t, test_prob),
          4
      ))

print("PR-AUC:",
      round(
          average_precision_score(y_test_t, test_prob),
          4
      ))
```

```python
# ============================================================
# PTXPHISH — BUILD KNOWN THREAT ADDRESS TABLE
# ============================================================

import pandas as pd
import re

initial_path = "/content/PTXPhish/dataset/InitialAddress.xlsx"

raw = pd.read_excel(
    initial_path,
    header=None
)

ADDRESS_RE = re.compile(
    r"0x[a-fA-F0-9]{40}"
)

records = []

# The first row contains attack-type block names.
# Each block is roughly:
# [attack_type, Tx_total, Type, blank]

current_attack = None
col = 0

while col < raw.shape[1]:

    header = raw.iloc[0, col]

    if pd.notna(header):
        current_attack = str(header).strip()

    # Data starts from row 1
    for row in range(1, raw.shape[0]):

        value = raw.iloc[row, col]

        if pd.isna(value):
            continue

        value = str(value).strip()

        match = ADDRESS_RE.fullmatch(value)

        if not match:
            continue

        tx_total = (
            raw.iloc[row, col + 1]
            if col + 1 < raw.shape[1]
            else None
        )

        tx_type = (
            raw.iloc[row, col + 2]
            if col + 2 < raw.shape[1]
            else None
        )

        records.append({
            "address": match.group(0).lower(),
            "attack_type": current_attack,
            "tx_total": pd.to_numeric(
                tx_total,
                errors="coerce"
            ),
            "tx_type": pd.to_numeric(
                tx_type,
                errors="coerce"
            )
        })

    col += 4


threat_df = pd.DataFrame(records)

# Remove empty / malformed records
threat_df = threat_df.dropna(
    subset=["address"]
)

# Remove duplicates
threat_df = threat_df.drop_duplicates()

print("=" * 70)
print("KNOWN EVM THREAT ADDRESSES")
print("=" * 70)

print("Records:", len(threat_df))
print(
    "Unique addresses:",
    threat_df["address"].nunique()
)

print("\nAttack categories:")
print(
    threat_df["attack_type"]
    .value_counts()
)

print("\nSample:")
display(threat_df.head(20))
```

```python
# ============================================================
# THREAT ADDRESS SUMMARY
# ============================================================

address_summary = (
    threat_df
    .groupby("address")
    .agg(
        attack_types=("attack_type", lambda x: list(sorted(set(x)))),
        attack_count=("attack_type", "nunique"),
        max_tx_total=("tx_total", "max"),
        min_tx_total=("tx_total", "min"),
    )
    .reset_index()
)

address_summary["known_malicious"] = 1

print("Unique known malicious addresses:",
      len(address_summary))

display(address_summary.head(20))
```

```python
# ============================================================
# PTXPHISH — BUILD LABELED TRANSACTION TABLE
# ============================================================

import pandas as pd
import re

ptx_path = "/content/PTXPhish/dataset/PTXPHISH.xlsx"

ptx = pd.read_excel(
    ptx_path,
    header=None
)

TX_RE = re.compile(r"^0x[a-fA-F0-9]{64}$")

# Row 3 contains category names.
# Data begins at row 4.
header_row = 3

records = []

for col in range(ptx.shape[1]):

    category = ptx.iloc[header_row, col]

    if pd.isna(category):
        continue

    category = str(category).strip()

    # Ignore the "Source" columns
    if category.lower() == "source":
        continue

    for row in range(header_row + 1, ptx.shape[0]):

        value = ptx.iloc[row, col]

        if pd.isna(value):
            continue

        value = str(value).strip()

        # Only accept COMPLETE 64-hex transaction hashes
        if TX_RE.fullmatch(value):

            is_benign = category.lower() in {
                "benign kol",
                "benign developer"
            }

            records.append({
                "tx_hash": value.lower(),
                "category": category,
                "label": 0 if is_benign else 1
            })

ptx_transactions = pd.DataFrame(records)

print("=" * 70)
print("PTXPHISH TRANSACTION DATASET")
print("=" * 70)

print("Raw labeled records:", len(ptx_transactions))
print("Unique transaction hashes:",
      ptx_transactions["tx_hash"].nunique())

print("\nLabel distribution:")
print(ptx_transactions["label"].value_counts())

print("\nCategory distribution:")
print(ptx_transactions["category"].value_counts())

display(ptx_transactions.head(20))
```

```python
# ============================================================
# PTXPHISH — FULL ETHEREUM TRANSACTION ENRICHMENT
# CHECKPOINTED + RETRY
# ============================================================

import requests
import pandas as pd
import time
import os
from pathlib import Path

RPC_URL = "https://ethereum-rpc.publicnode.com"

OUTPUT_DIR = Path("/content/ptxphish_enriched")
OUTPUT_DIR.mkdir(exist_ok=True)

CHECKPOINT_FILE = OUTPUT_DIR / "transactions.parquet"
FAILED_FILE = OUTPUT_DIR / "failed_hashes.txt"

BATCH_SIZE = 10
SLEEP_SECONDS = 0.15
MAX_RETRIES = 4

all_hashes = (
    ptx_transactions["tx_hash"]
    .astype(str)
    .str.lower()
    .drop_duplicates()
    .tolist()
)

print("Total transaction hashes:", len(all_hashes))


def fetch_batch(hashes):

    payload = [
        {
            "jsonrpc": "2.0",
            "id": i,
            "method": "eth_getTransactionByHash",
            "params": [tx_hash]
        }
        for i, tx_hash in enumerate(hashes)
    ]

    for attempt in range(MAX_RETRIES):

        try:

            r = requests.post(
                RPC_URL,
                json=payload,
                headers={
                    "Content-Type": "application/json"
                },
                timeout=60
            )

            r.raise_for_status()

            body = r.json()

            if not isinstance(body, list):
                raise RuntimeError(
                    f"Unexpected RPC response: {body}"
                )

            return body

        except Exception as exc:

            print(
                f"Batch error "
                f"(attempt {attempt + 1}/{MAX_RETRIES}):",
                exc
            )

            if attempt < MAX_RETRIES - 1:
                time.sleep(2 ** attempt)

    return None


def parse_response(body):

    rows = []
    failed = []

    for item in body or []:

        if not isinstance(item, dict):
            continue

        item_id = item.get("id")

        if "error" in item:
            failed.append(item_id)
            continue

        tx = item.get("result")

        if tx is None:
            failed.append(item_id)
            continue

        try:
            rows.append({
                "tx_hash": tx.get("hash"),
                "block_number": (
                    int(tx["blockNumber"], 16)
                    if tx.get("blockNumber")
                    else None
                ),
                "from": tx.get("from"),
                "to": tx.get("to"),
                "gas": (
                    int(tx["gas"], 16)
                    if tx.get("gas")
                    else None
                ),
                "gas_price": (
                    int(tx["gasPrice"], 16)
                    if tx.get("gasPrice")
                    else None
                ),
                "value_wei": (
                    int(tx["value"], 16)
                    if tx.get("value")
                    else 0
                ),
                "nonce": (
                    int(tx["nonce"], 16)
                    if tx.get("nonce")
                    else None
                ),
                "input": tx.get("input", "")
            })

        except Exception:
            failed.append(item_id)

    return rows, failed


# ------------------------------------------------------------
# Resume from existing checkpoint
# ------------------------------------------------------------

if CHECKPOINT_FILE.exists():

    enriched = pd.read_parquet(
        CHECKPOINT_FILE
    )

    completed_hashes = set(
        enriched["tx_hash"]
        .astype(str)
        .str.lower()
    )

    print(
        "Existing checkpoint:",
        len(enriched)
    )

else:

    enriched = pd.DataFrame()

    completed_hashes = set()


remaining_hashes = [
    h for h in all_hashes
    if h not in completed_hashes
]

print(
    "Remaining:",
    len(remaining_hashes)
)


# ------------------------------------------------------------
# Download
# ------------------------------------------------------------

total_batches = (
    len(remaining_hashes) + BATCH_SIZE - 1
) // BATCH_SIZE


for batch_no, start in enumerate(
    range(0, len(remaining_hashes), BATCH_SIZE),
    start=1
):

    batch_hashes = remaining_hashes[
        start:start + BATCH_SIZE
    ]

    body = fetch_batch(batch_hashes)

    if body is None:
        print(
            f"Skipping batch {batch_no}/{total_batches}"
        )
        continue

    rows, failed_ids = parse_response(body)

    if rows:

        batch_df = pd.DataFrame(rows)

        enriched = pd.concat(
            [enriched, batch_df],
            ignore_index=True
        )

    # --------------------------------------------------------
    # checkpoint every 25 batches
    # --------------------------------------------------------

    if batch_no % 25 == 0:

        enriched = (
            enriched
            .drop_duplicates("tx_hash")
            .reset_index(drop=True)
        )

        enriched.to_parquet(
            CHECKPOINT_FILE,
            index=False
        )

        print(
            f"[{batch_no}/{total_batches}] "
            f"saved {len(enriched)} transactions"
        )

    time.sleep(SLEEP_SECONDS)


# Final save
enriched = (
    enriched
    .drop_duplicates("tx_hash")
    .reset_index(drop=True)
)

enriched.to_parquet(
    CHECKPOINT_FILE,
    index=False
)

print("\n" + "=" * 70)
print("ENRICHMENT COMPLETE")
print("=" * 70)

print("Retrieved:", len(enriched))
print("Expected :", len(all_hashes))
print("Saved to :", CHECKPOINT_FILE)
```

```python
# ============================================================
# EVM TRANSACTION MODEL — FEATURE ENGINEERING
# ============================================================

import pandas as pd
import numpy as np
import re

df = ptx_ml.copy()

# ------------------------------------------------------------
# Clean numeric blockchain fields
# ------------------------------------------------------------

for col in [
    "block_number",
    "gas",
    "gas_price",
    "value_wei",
    "nonce"
]:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    ).fillna(0)


# ------------------------------------------------------------
# Input data features
# ------------------------------------------------------------

df["input"] = df["input"].fillna("").astype(str)

# Function selector = first 4 bytes = 10 characters including 0x
df["method_selector"] = (
    df["input"]
    .str[:10]
    .str.lower()
)

# Encode selector without using raw hexadecimal as a category
df["method_selector_num"] = (
    df["method_selector"]
    .apply(
        lambda x:
        int(x[2:], 16)
        if re.fullmatch(r"0x[0-9a-f]{8}", x)
        else 0
    )
)

df["input_length"] = (
    df["input"]
    .str.len()
)

df["input_word_count"] = (
    np.maximum(
        (df["input_length"] - 10) / 64,
        0
    )
)


# ------------------------------------------------------------
# Transaction behavior
# ------------------------------------------------------------

df["value_is_zero"] = (
    df["value_wei"] == 0
).astype(int)

df["has_input"] = (
    df["input_length"] > 10
).astype(int)

df["is_contract_call"] = (
    (df["to"].notna()) &
    (df["input_length"] > 10)
).astype(int)


# ------------------------------------------------------------
# Human-readable monetary units
# ------------------------------------------------------------

df["value_eth"] = (
    df["value_wei"] /
    10**18
)

df["gas_price_gwei"] = (
    df["gas_price"] /
    10**9
)


# ------------------------------------------------------------
# Log-scaled features
# Helps deal with extremely large blockchain values
# ------------------------------------------------------------

for col in [
    "gas",
    "gas_price_gwei",
    "value_eth",
    "nonce",
    "input_length",
    "input_word_count"
]:
    df[f"{col}_log"] = np.log1p(
        np.maximum(
            df[col].astype(float),
            0
        )
    )


# ------------------------------------------------------------
# Keep only ML-safe numerical features
# ------------------------------------------------------------

feature_columns = [
    "block_number",
    "gas",
    "gas_price_gwei",
    "value_eth",
    "nonce",

    "input_length",
    "input_word_count",
    "method_selector_num",

    "value_is_zero",
    "has_input",
    "is_contract_call",

    "gas_log",
    "gas_price_gwei_log",
    "value_eth_log",
    "nonce_log",
    "input_length_log",
    "input_word_count_log"
]

X_evm = df[feature_columns].copy()
y_evm = df["label"].astype(int)

# Clean
X_evm = (
    X_evm
    .replace([np.inf, -np.inf], np.nan)
    .fillna(0)
)

print("=" * 70)
print("EVM FEATURE MATRIX")
print("=" * 70)

print("Rows:", len(X_evm))
print("Features:", len(feature_columns))

print("\nLabel distribution:")
print(y_evm.value_counts())

print("\nFeatures:")
print(feature_columns)

display(X_evm.head())
```

```python
# ============================================================
# EVM TRANSACTION XGBOOST
# ============================================================

from sklearn.model_selection import train_test_split
from xgboost import XGBClassifier

X_train_evm, X_test_evm, y_train_evm, y_test_evm = train_test_split(
    X_evm,
    y_evm,
    test_size=0.25,
    stratify=y_evm,
    random_state=42
)

negative = int((y_train_evm == 0).sum())
positive = int((y_train_evm == 1).sum())

evm_weight = negative / positive

print("Training:", X_train_evm.shape)
print("Testing :", X_test_evm.shape)
print("Class weight:", evm_weight)


evm_model = XGBClassifier(
    n_estimators=500,
    max_depth=5,
    learning_rate=0.05,

    subsample=0.85,
    colsample_bytree=0.85,

    objective="binary:logistic",
    eval_metric="aucpr",

    scale_pos_weight=evm_weight,

    tree_method="hist",
    n_jobs=-1,
    random_state=42
)

print("\nTraining EVM transaction model...")

evm_model.fit(
    X_train_evm,
    y_train_evm,
    eval_set=[
        (X_test_evm, y_test_evm)
    ],
    verbose=False
)

print("\nEVM MODEL COMPLETE")
```

```python
# ============================================================
# EVM MODEL — EVALUATION
# ============================================================

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score,
    average_precision_score
)

evm_prob = evm_model.predict_proba(
    X_test_evm
)[:, 1]

evm_pred = (
    evm_prob >= 0.50
).astype(int)

print("=" * 70)
print("EVM TRANSACTION MODEL RESULTS")
print("=" * 70)

print(
    classification_report(
        y_test_evm,
        evm_pred,
        target_names=[
            "BENIGN",
            "PHISHING"
        ],
        digits=4
    )
)

print("\nCONFUSION MATRIX:")
print(
    confusion_matrix(
        y_test_evm,
        evm_pred
    )
)

print(
    "\nROC-AUC:",
    round(
        roc_auc_score(
            y_test_evm,
            evm_prob
        ),
        4
    )
)

print(
    "PR-AUC:",
    round(
        average_precision_score(
            y_test_evm,
            evm_prob
        ),
        4
    )
)
```

```python
# ============================================================
# EVM MODEL V2 — PAYLOAD-AWARE FEATURES
# ============================================================

import pandas as pd
import numpy as np
import re

evm_v2 = ptx_ml.copy()

# ------------------------------------------------------------
# Numeric conversion
# ------------------------------------------------------------

for col in [
    "gas",
    "gas_price",
    "value_wei",
    "nonce",
    "block_number"
]:
    evm_v2[col] = pd.to_numeric(
        evm_v2[col],
        errors="coerce"
    ).fillna(0)


# ------------------------------------------------------------
# Calldata / payload features
# ------------------------------------------------------------

evm_v2["input"] = (
    evm_v2["input"]
    .fillna("")
    .astype(str)
    .str.lower()
)

# 4-byte method selector
evm_v2["selector"] = (
    evm_v2["input"]
    .str[:10]
)

# Basic calldata characteristics
evm_v2["input_length"] = evm_v2["input"].str.len()

evm_v2["calldata_hex_length"] = (
    evm_v2["input_length"] - 2
)

evm_v2["input_word_count"] = np.maximum(
    (evm_v2["input_length"] - 10) / 64,
    0
)

# Count zero bytes in calldata
evm_v2["zero_byte_count"] = (
    evm_v2["input"]
    .str.count("00")
)

# Count non-zero bytes
evm_v2["nonzero_byte_count"] = (
    evm_v2["input"]
    .str.count(r"[1-9a-f]")
)


# ------------------------------------------------------------
# Transaction structure
# ------------------------------------------------------------

evm_v2["value_is_zero"] = (
    evm_v2["value_wei"] == 0
).astype(int)

evm_v2["has_calldata"] = (
    evm_v2["input_length"] > 10
).astype(int)

evm_v2["is_contract_call"] = (
    evm_v2["to"].notna() &
    (evm_v2["input_length"] > 10)
).astype(int)


# ------------------------------------------------------------
# Unit conversions
# ------------------------------------------------------------

evm_v2["value_eth"] = (
    evm_v2["value_wei"] / 10**18
)

evm_v2["gas_price_gwei"] = (
    evm_v2["gas_price"] / 10**9
)


# ------------------------------------------------------------
# Log transforms
# ------------------------------------------------------------

for col in [
    "gas",
    "gas_price_gwei",
    "value_eth",
    "nonce",
    "input_length",
    "input_word_count",
    "zero_byte_count",
    "nonzero_byte_count"
]:

    evm_v2[f"{col}_log"] = np.log1p(
        np.maximum(
            evm_v2[col].astype(float),
            0
        )
    )


# ------------------------------------------------------------
# Selector family features
# ------------------------------------------------------------

# Common Ethereum function selectors.
# These are generic ABI selectors, not PTXPhish labels.

SELECTOR_MAP = {
    "approve": {
        "0x095ea7b3"
    },

    "transfer": {
        "0xa9059cbb"
    },

    "transferFrom": {
        "0x23b872dd"
    },

    "setApprovalForAll": {
        "0xa22cb465"
    },

    "permit": {
        "0xd505accf",
        "0x8fcbaf0c"
    },

    "safeTransferFrom": {
        "0x42842e0e",
        "0xb88d4fde"
    }
}

for name, selectors in SELECTOR_MAP.items():

    evm_v2[f"selector_{name}"] = (
        evm_v2["selector"]
        .isin(selectors)
        .astype(int)
    )

evm_v2["known_selector"] = (
    evm_v2[
        [
            f"selector_{name}"
            for name in SELECTOR_MAP
        ]
    ].max(axis=1)
)

evm_v2["unknown_selector"] = (
    1 - evm_v2["known_selector"]
)


# ------------------------------------------------------------
# Final feature list
# ------------------------------------------------------------

feature_columns_v2 = [
    "gas",
    "gas_price_gwei",
    "value_eth",
    "nonce",

    "input_length",
    "input_word_count",
    "zero_byte_count",
    "nonzero_byte_count",

    "value_is_zero",
    "has_calldata",
    "is_contract_call",

    "gas_log",
    "gas_price_gwei_log",
    "value_eth_log",
    "nonce_log",
    "input_length_log",
    "input_word_count_log",
    "zero_byte_count_log",
    "nonzero_byte_count_log",

    "selector_approve",
    "selector_transfer",
    "selector_transferFrom",
    "selector_setApprovalForAll",
    "selector_permit",
    "selector_safeTransferFrom",

    "known_selector",
    "unknown_selector"
]

X_evm_v2 = (
    evm_v2[feature_columns_v2]
    .replace([np.inf, -np.inf], np.nan)
    .fillna(0)
)

y_evm_v2 = evm_v2["label"].astype(int)

print("=" * 70)
print("EVM V2 FEATURE MATRIX")
print("=" * 70)

print("Rows:", len(X_evm_v2))
print("Features:", len(feature_columns_v2))

print("\nLabels:")
print(y_evm_v2.value_counts())

display(X_evm_v2.head())
```

```python
# ============================================================
# EVM V2 — SENDER-GROUPED TRAIN / TEST SPLIT
# ============================================================

from sklearn.model_selection import GroupShuffleSplit
from xgboost import XGBClassifier

groups = (
    evm_v2["from"]
    .fillna("")
    .astype(str)
    .str.lower()
)

splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.25,
    random_state=42
)

train_idx, test_idx = next(
    splitter.split(
        X_evm_v2,
        y_evm_v2,
        groups=groups
    )
)

X_train_evm_v2 = X_evm_v2.iloc[train_idx]
X_test_evm_v2 = X_evm_v2.iloc[test_idx]

y_train_evm_v2 = y_evm_v2.iloc[train_idx]
y_test_evm_v2 = y_evm_v2.iloc[test_idx]

print("Train:", X_train_evm_v2.shape)
print("Test :", X_test_evm_v2.shape)

print("\nTrain labels:")
print(y_train_evm_v2.value_counts())

print("\nTest labels:")
print(y_test_evm_v2.value_counts())
```

```python
# ============================================================
# TRAIN EVM MODEL V2
# ============================================================

negative = int(
    (y_train_evm_v2 == 0).sum()
)

positive = int(
    (y_train_evm_v2 == 1).sum()
)

weight_v2 = negative / positive

model_evm_v2 = XGBClassifier(
    n_estimators=500,
    max_depth=5,
    learning_rate=0.05,

    subsample=0.85,
    colsample_bytree=0.85,

    objective="binary:logistic",
    eval_metric="aucpr",

    scale_pos_weight=weight_v2,

    tree_method="hist",
    n_jobs=-1,
    random_state=42
)

print("Training EVM V2...")

model_evm_v2.fit(
    X_train_evm_v2,
    y_train_evm_v2,
    eval_set=[
        (X_test_evm_v2, y_test_evm_v2)
    ],
    verbose=False
)

print("EVM V2 COMPLETE")
```

```python
# ============================================================
# EVM V2 — EVALUATION
# ============================================================

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score,
    average_precision_score
)

v2_prob = model_evm_v2.predict_proba(
    X_test_evm_v2
)[:, 1]

v2_pred = (
    v2_prob >= 0.50
).astype(int)

print("=" * 70)
print("EVM V2 — GROUPED TEST")
print("=" * 70)

print(
    classification_report(
        y_test_evm_v2,
        v2_pred,
        target_names=[
            "BENIGN",
            "PHISHING"
        ],
        digits=4
    )
)

print("\nConfusion matrix:")
print(
    confusion_matrix(
        y_test_evm_v2,
        v2_pred
    )
)

print(
    "\nROC-AUC:",
    round(
        roc_auc_score(
            y_test_evm_v2,
            v2_prob
        ),
        4
    )
)

print(
    "PR-AUC:",
    round(
        average_precision_score(
            y_test_evm_v2,
            v2_prob
        ),
        4
    )
)
```

```python
# ============================================================
# EVM MODEL V3 — WALLET GRAPH FEATURE EXTRACTION
# ============================================================

import os
import pandas as pd
import numpy as np

BASE = "/content/ethereum_partial/Ethereum Partial Transaction Dataset"


def load_addr_map(path):
    """
    addr2Idx format:
        ethereum_address,node_id
    """
    rows = []

    with open(
        path,
        "r",
        encoding="utf-8",
        errors="replace"
    ) as f:

        for line in f:

            line = line.strip()

            if not line:
                continue

            parts = line.split(",")

            if len(parts) < 2:
                continue

            address = parts[0].strip().lower()

            try:
                node_id = int(float(parts[1]))
            except Exception:
                continue

            rows.append(
                (node_id, address)
            )

    return pd.DataFrame(
        rows,
        columns=["node_id", "address"]
    ).drop_duplicates("node_id")


def extract_graph_features(group_name):

    gid = group_name[-1]
    group_dir = os.path.join(BASE, group_name)

    # --------------------------------------------------------
    # Address mapping
    # --------------------------------------------------------

    addr_path = os.path.join(
        group_dir,
        "addr2Idx.txt"
    )

    addr_map = load_addr_map(addr_path)

    # --------------------------------------------------------
    # Find edge list
    # --------------------------------------------------------

    edge_file = None

    for suffix in ["0.5", "0.45"]:

        candidate = os.path.join(
            group_dir,
            f"LPsubG{gid}_{suffix}_TransEdgelist.txt"
        )

        if os.path.exists(candidate):
            edge_file = candidate
            break

    if edge_file is None:
        raise FileNotFoundError(
            f"No edge file found for {group_name}"
        )

    print(f"\n{group_name}")
    print("Edge file:", edge_file)
    print("Addresses:", len(addr_map))

    # --------------------------------------------------------
    # Read edge list
    # --------------------------------------------------------

    edges = pd.read_csv(
        edge_file,
        header=None,
        names=[
            "from_id",
            "to_id",
            "value",
            "timestamp"
        ]
    )

    edges["from_id"] = pd.to_numeric(
        edges["from_id"],
        errors="coerce"
    )

    edges["to_id"] = pd.to_numeric(
        edges["to_id"],
        errors="coerce"
    )

    edges["value"] = pd.to_numeric(
        edges["value"],
        errors="coerce"
    ).fillna(0)

    edges["timestamp"] = pd.to_numeric(
        edges["timestamp"],
        errors="coerce"
    )

    edges = edges.dropna(
        subset=[
            "from_id",
            "to_id"
        ]
    )

    print("Transactions:", len(edges))

    # --------------------------------------------------------
    # Outgoing features
    # --------------------------------------------------------

    outgoing = (
        edges
        .groupby("from_id")
        .agg(
            out_tx_count=("to_id", "size"),
            out_unique_counterparties=("to_id", "nunique"),
            out_total_value=("value", "sum"),
            out_mean_value=("value", "mean"),
            out_max_value=("value", "max"),
            out_first_time=("timestamp", "min"),
            out_last_time=("timestamp", "max")
        )
        .reset_index()
        .rename(
            columns={"from_id": "node_id"}
        )
    )

    # --------------------------------------------------------
    # Incoming features
    # --------------------------------------------------------

    incoming = (
        edges
        .groupby("to_id")
        .agg(
            in_tx_count=("from_id", "size"),
            in_unique_counterparties=("from_id", "nunique"),
            in_total_value=("value", "sum"),
            in_mean_value=("value", "mean"),
            in_max_value=("value", "max"),
            in_first_time=("timestamp", "min"),
            in_last_time=("timestamp", "max")
        )
        .reset_index()
        .rename(
            columns={"to_id": "node_id"}
        )
    )

    # --------------------------------------------------------
    # Overall participation
    # --------------------------------------------------------

    all_nodes = pd.DataFrame({
        "node_id": pd.concat(
            [
                edges["from_id"],
                edges["to_id"]
            ],
            ignore_index=True
        ).unique()
    })

    features = (
        all_nodes
        .merge(
            outgoing,
            on="node_id",
            how="left"
        )
        .merge(
            incoming,
            on="node_id",
            how="left"
        )
        .fillna(0)
    )

    # --------------------------------------------------------
    # Derived behavior
    # --------------------------------------------------------

    features["total_tx_count"] = (
        features["in_tx_count"] +
        features["out_tx_count"]
    )

    features["total_unique_counterparties"] = (
        features["in_unique_counterparties"] +
        features["out_unique_counterparties"]
    )

    features["total_value"] = (
        features["in_total_value"] +
        features["out_total_value"]
    )

    features["net_flow"] = (
        features["in_total_value"] -
        features["out_total_value"]
    )

    features["flow_ratio"] = (
        features["out_total_value"] /
        (
            features["in_total_value"] +
            1e-12
        )
    )

    features["activity_start"] = np.minimum(
        features["in_first_time"],
        features["out_first_time"]
    )

    # Replace zeros produced by missing groups
    features["activity_start"] = (
        features["activity_start"]
        .replace(0, np.nan)
    )

    features["activity_end"] = np.maximum(
        features["in_last_time"],
        features["out_last_time"]
    )

    features["activity_span"] = (
        features["activity_end"] -
        features["activity_start"]
    ).fillna(0)

    features["fan_in"] = (
        features["in_unique_counterparties"]
    )

    features["fan_out"] = (
        features["out_unique_counterparties"]
    )

    features["degree"] = (
        features["fan_in"] +
        features["fan_out"]
    )

    # --------------------------------------------------------
    # Log features
    # --------------------------------------------------------

    for col in [
        "total_tx_count",
        "total_value",
        "in_total_value",
        "out_total_value",
        "degree",
        "fan_in",
        "fan_out",
        "activity_span"
    ]:

        features[f"{col}_log"] = np.log1p(
            np.maximum(
                features[col].astype(float),
                0
            )
        )

    # --------------------------------------------------------
    # Map node IDs -> Ethereum addresses
    # --------------------------------------------------------

    features = features.merge(
        addr_map,
        on="node_id",
        how="left"
    )

    return features


# ============================================================
# BUILD G1 / G2 / G3
# ============================================================

graph_feature_frames = []

for group in [
    "EthereumG1",
    "EthereumG2",
    "EthereumG3"
]:

    try:

        f = extract_graph_features(group)

        f["source_graph"] = group

        graph_feature_frames.append(f)

    except Exception as exc:

        print(
            f"{group} failed:",
            type(exc).__name__,
            exc
        )


wallet_graph_features = pd.concat(
    graph_feature_frames,
    ignore_index=True
)

# Same address may appear across graph snapshots.
wallet_graph_features = (
    wallet_graph_features
    .dropna(subset=["address"])
    .drop_duplicates(
        subset=[
            "source_graph",
            "address"
        ]
    )
)

print("\n" + "=" * 70)
print("GRAPH FEATURE DATASET")
print("=" * 70)

print(
    "Rows:",
    len(wallet_graph_features)
)

print(
    "Unique addresses:",
    wallet_graph_features["address"].nunique()
)

print(
    "Graphs:",
    wallet_graph_features["source_graph"].value_counts()
)

display(
    wallet_graph_features.head()
)
```

```python
# ============================================================
# REBUILD EVM THREAT INTELLIGENCE AFTER RUNTIME RESET
# ============================================================

import pandas as pd
import re

# ------------------------------------------------------------
# RiskProp
# ------------------------------------------------------------

riskprop_path = "/content/RiskProp/data/all_hash_0_hop_803_labels.csv"

rp = pd.read_csv(riskprop_path)

riskprop_threats = pd.DataFrame({
    "address": (
        rp.loc[rp["label"] == 1, "Address"]
        .astype(str)
        .str.lower()
        .str.strip()
    ),
    "source": "RiskProp",
    "threat_type": "phish-hack"
})

# ------------------------------------------------------------
# PTXPhish InitialAddress
# ------------------------------------------------------------

initial_path = "/content/PTXPhish/dataset/InitialAddress.xlsx"

initial = pd.read_excel(
    initial_path,
    header=None
)

ADDRESS_RE = re.compile(
    r"^0x[a-fA-F0-9]{40}$"
)

records = []

# Row 0 contains attack category names.
# Each category occupies roughly 4 columns:
# [category, Tx_total, Type, blank]

for col in range(initial.shape[1]):

    category = initial.iloc[0, col]

    if pd.isna(category):
        continue

    category = str(category).strip()

    for row in range(1, initial.shape[0]):

        value = initial.iloc[row, col]

        if pd.isna(value):
            continue

        value = str(value).strip()

        if not ADDRESS_RE.fullmatch(value):
            continue

        records.append({
            "address": value.lower(),
            "attack_type": category
        })

ptx_threats = pd.DataFrame(records)

ptx_threats = (
    ptx_threats
    .drop_duplicates()
)

# ------------------------------------------------------------
# Unified threat intelligence
# ------------------------------------------------------------

ptx_intel = (
    ptx_threats
    .groupby("address")["attack_type"]
    .apply(lambda x: ", ".join(sorted(set(x))))
    .reset_index()
)

ptx_intel["source"] = "PTXPhish"
ptx_intel["threat_type"] = ptx_intel["attack_type"]

ptx_intel = ptx_intel[
    ["address", "source", "threat_type"]
]

evm_threat_intel = pd.concat(
    [
        riskprop_threats,
        ptx_intel
    ],
    ignore_index=True
)

evm_threat_intel = (
    evm_threat_intel
    .drop_duplicates(
        subset=["address", "source"]
    )
)

print("=" * 70)
print("EVM THREAT INTELLIGENCE")
print("=" * 70)

print(
    "Records:",
    len(evm_threat_intel)
)

print(
    "Unique addresses:",
    evm_threat_intel["address"].nunique()
)

print("\nSources:")
print(
    evm_threat_intel["source"].value_counts()
)
```

```python
# ============================================================
# MODEL V3 — EVM WALLET ANOMALY DETECTOR
# ============================================================

import numpy as np
import pandas as pd
import joblib
import os

from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import RobustScaler

# ------------------------------------------------------------
# Select numerical behavioral features
# ------------------------------------------------------------

exclude_cols = {
    "node_id",
    "address",
    "source_graph",
    "known_threat"
}

anomaly_features = [
    c for c in wallet_graph_features.columns
    if c not in exclude_cols
    and pd.api.types.is_numeric_dtype(
        wallet_graph_features[c]
    )
]

print("Anomaly features:", len(anomaly_features))
print(anomaly_features)

X_anomaly = (
    wallet_graph_features[anomaly_features]
    .replace([np.inf, -np.inf], np.nan)
    .fillna(0)
)

# ------------------------------------------------------------
# Robust scaling
# ------------------------------------------------------------

scaler_v3 = RobustScaler()

X_scaled = scaler_v3.fit_transform(
    X_anomaly
)

# ------------------------------------------------------------
# Isolation Forest
# contamination is NOT "fraud percentage".
# It is an estimate of how much unusual behavior
# we want the detector to flag.
# ------------------------------------------------------------

iso_v3 = IsolationForest(
    n_estimators=400,
    max_samples="auto",
    contamination="auto",
    random_state=42,
    n_jobs=-1
)

print("\nTraining Isolation Forest...")

iso_v3.fit(X_scaled)

# ------------------------------------------------------------
# Scores
#
# sklearn's decision_function:
# larger = more normal
#
# We invert it so:
# larger anomaly_score = more anomalous
# ------------------------------------------------------------

wallet_graph_features["anomaly_raw"] = (
    -iso_v3.decision_function(X_scaled)
)

# Normalize approximately to 0-100
raw = wallet_graph_features["anomaly_raw"]

lo = raw.quantile(0.01)
hi = raw.quantile(0.99)

wallet_graph_features["anomaly_score"] = (
    (raw - lo) /
    (hi - lo)
).clip(0, 1) * 100

print("\nAnomaly model complete.")

print(
    "\nScore statistics:"
)

print(
    wallet_graph_features[
        "anomaly_score"
    ].describe()
)

display(
    wallet_graph_features[
        [
            "address",
            "anomaly_score"
        ] + anomaly_features[:5]
    ]
    .sort_values(
        "anomaly_score",
        ascending=False
    )
    .head(20)
)
```

```python
# ============================================================
# MODEL V5 — EXPLAINABLE GRAPH RISK
# ============================================================

import numpy as np
import pandas as pd

graph_risk = wallet_profile.copy()

# ------------------------------------------------------------
# Helper: percentile rank
# ------------------------------------------------------------

def percentile_score(series):
    return (
        series
        .rank(pct=True, method="average")
        .fillna(0)
        .clip(0, 1)
        * 100
    )

# ------------------------------------------------------------
# Aggregate cross-snapshot features
# ------------------------------------------------------------

# Use the largest observed structural activity
graph_risk["degree_signal"] = percentile_score(
    graph_risk["degree_max"]
)

graph_risk["fan_in_signal"] = percentile_score(
    graph_risk["fan_in_max"]
)

graph_risk["fan_out_signal"] = percentile_score(
    graph_risk["fan_out_max"]
)

graph_risk["tx_signal"] = percentile_score(
    graph_risk["total_tx_count_max"]
)

graph_risk["counterparty_signal"] = percentile_score(
    graph_risk["total_unique_counterparties_max"]
)

# ------------------------------------------------------------
# Flow imbalance
# ------------------------------------------------------------

# Avoid the huge values caused by zero inbound flow.
flow_ratio = (
    graph_risk["flow_ratio_mean"]
    .replace([np.inf, -np.inf], np.nan)
    .fillna(0)
)

graph_risk["flow_imbalance_signal"] = percentile_score(
    np.abs(
        np.log1p(
            np.abs(flow_ratio)
        )
    )
)

# ------------------------------------------------------------
# High activity + high fan-out is particularly interesting
# for wallet-network risk.
# ------------------------------------------------------------

graph_risk["fanout_activity_signal"] = (
    graph_risk["fan_out_signal"] *
    graph_risk["tx_signal"]
    / 100
)

# ------------------------------------------------------------
# Final explainable graph score
# ------------------------------------------------------------

graph_risk["graph_risk_score"] = (
      0.20 * graph_risk["degree_signal"]
    + 0.15 * graph_risk["fan_in_signal"]
    + 0.20 * graph_risk["fan_out_signal"]
    + 0.15 * graph_risk["counterparty_signal"]
    + 0.10 * graph_risk["tx_signal"]
    + 0.10 * graph_risk["flow_imbalance_signal"]
    + 0.10 * graph_risk["fanout_activity_signal"]
)

graph_risk["graph_risk_score"] = (
    graph_risk["graph_risk_score"]
    .clip(0, 100)
)

print("=" * 70)
print("GRAPH RISK V5")
print("=" * 70)

print(
    graph_risk["graph_risk_score"]
    .describe()
)

display(
    graph_risk[
        [
            "address",
            "graph_risk_score",
            "degree_max",
            "fan_in_max",
            "fan_out_max",
            "total_tx_count_max",
            "total_unique_counterparties_max"
        ]
    ]
    .sort_values(
        "graph_risk_score",
        ascending=False
    )
    .head(25)
)
```

```python
# ============================================================
# GRAPH RISK — EXPLANATIONS
# ============================================================

def graph_reasons(row):

    reasons = []

    if row["fan_out_signal"] >= 90:
        reasons.append("very high outbound connectivity")

    if row["fan_in_signal"] >= 90:
        reasons.append("very high inbound connectivity")

    if row["counterparty_signal"] >= 90:
        reasons.append("large counterparty network")

    if row["degree_signal"] >= 90:
        reasons.append("high graph degree")

    if row["tx_signal"] >= 90:
        reasons.append("very high transaction activity")

    if row["flow_imbalance_signal"] >= 90:
        reasons.append("strongly imbalanced value flow")

    return reasons


graph_risk["graph_reasons"] = (
    graph_risk.apply(
        graph_reasons,
        axis=1
    )
)

display(
    graph_risk[
        [
            "address",
            "graph_risk_score",
            "graph_reasons"
        ]
    ]
    .sort_values(
        "graph_risk_score",
        ascending=False
    )
    .head(20)
)
```

```python
# ============================================================
# FINAL RISK FUSION ENGINE V1
# ============================================================

import numpy as np
import pandas as pd


def clamp_score(x):
    """Force a value into the 0-100 range."""
    try:
        return float(np.clip(float(x), 0, 100))
    except (TypeError, ValueError):
        return 0.0


def risk_level(score):
    score = clamp_score(score)

    if score >= 80:
        return "CRITICAL"
    elif score >= 60:
        return "HIGH"
    elif score >= 35:
        return "MEDIUM"
    else:
        return "LOW"


def fuse_risk(
    tx_risk=0,
    anomaly_score=0,
    graph_risk=0,
    threat_score=0,
    threat_sources=None,
    threat_types=None,
):
    """
    Transparent prototype risk fusion.

    We deliberately keep the weights explicit so they can
    later be calibrated against a properly labeled validation
    corpus.
    """

    tx_risk = clamp_score(tx_risk)
    anomaly_score = clamp_score(anomaly_score)
    graph_risk = clamp_score(graph_risk)
    threat_score = clamp_score(threat_score)

    # Initial prototype weights
    weights = {
        "transaction": 0.35,
        "anomaly": 0.20,
        "graph": 0.20,
        "threat_intel": 0.25,
    }

    weighted_score = (
        weights["transaction"] * tx_risk
        + weights["anomaly"] * anomaly_score
        + weights["graph"] * graph_risk
        + weights["threat_intel"] * threat_score
    )

    # Known malicious intelligence should never be diluted
    # by a low ML score.
    if threat_score >= 100:
        final_score = max(weighted_score, 95.0)
    else:
        final_score = weighted_score

    final_score = round(
        clamp_score(final_score),
        2
    )

    reasons = []

    if threat_score >= 100:
        reasons.append(
            "Address matched known threat intelligence"
        )

    if tx_risk >= 80:
        reasons.append(
            "Transaction behavior strongly resembles phishing activity"
        )
    elif tx_risk >= 60:
        reasons.append(
            "Transaction behavior shows elevated phishing risk"
        )

    if anomaly_score >= 80:
        reasons.append(
            "Wallet behavior is highly anomalous"
        )
    elif anomaly_score >= 60:
        reasons.append(
            "Wallet behavior is unusually structured"
        )

    if graph_risk >= 80:
        reasons.append(
            "Wallet occupies a high-risk graph position"
        )
    elif graph_risk >= 60:
        reasons.append(
            "Wallet shows elevated graph connectivity risk"
        )

    if threat_types:
        for threat_type in threat_types:
            reasons.append(
                f"Threat intelligence: {threat_type}"
            )

    return {
        "risk_score": final_score,
        "risk_level": risk_level(final_score),

        "signals": {
            "transaction_risk": round(tx_risk, 2),
            "anomaly_score": round(anomaly_score, 2),
            "graph_risk": round(graph_risk, 2),
            "threat_intelligence": round(threat_score, 2),
        },

        "weights": weights,

        "threat_sources": (
            sorted(set(threat_sources))
            if threat_sources
            else []
        ),

        "threat_types": (
            sorted(set(threat_types))
            if threat_types
            else []
        ),

        "reasons": reasons,
    }


print("Risk Fusion Engine V1 ready.")
```

```python
# ============================================================
# WALLET INVESTIGATION RISK FUNCTION
# ============================================================

def investigate_wallet_risk(
    address,
    tx_risk=0,
    anomaly_score=0,
    graph_risk=0,
):
    """
    Combine all available intelligence for one wallet.
    """

    address = str(
        address
    ).lower().strip()

    intel = threat_lookup.get(
        address,
        {
            "sources": [],
            "types": []
        }
    )

    is_known_threat = bool(
        intel["sources"]
    )

    threat_score = (
        100.0
        if is_known_threat
        else 0.0
    )

    result = fuse_risk(
        tx_risk=tx_risk,
        anomaly_score=anomaly_score,
        graph_risk=graph_risk,
        threat_score=threat_score,
        threat_sources=intel["sources"],
        threat_types=intel["types"],
    )

    result["address"] = address

    result["known_threat"] = (
        is_known_threat
    )

    return result


print("Wallet investigation function ready.")
```

```python
def build_evm_model_features(
    gas,
    gas_price,
    value_wei,
    nonce,
    input_data,
    to_address=None,
):
    import re
    import numpy as np

    input_data = (
        input_data
        if isinstance(input_data, str)
        else ""
    ).lower()

    gas = float(gas or 0)
    gas_price = float(gas_price or 0)
    value_wei = float(value_wei or 0)
    nonce = float(nonce or 0)

    input_length = len(input_data)

    input_word_count = max(
        (input_length - 10) / 64,
        0
    )

    zero_byte_count = input_data.count("00")

    nonzero_byte_count = len(
        re.findall(
            r"[1-9a-f]",
            input_data
        )
    )

    value_eth = (
        value_wei / 10**18
    )

    gas_price_gwei = (
        gas_price / 10**9
    )

    selector = input_data[:10]

    SELECTOR_MAP = {
        "approve": {"0x095ea7b3"},
        "transfer": {"0xa9059cbb"},
        "transferFrom": {"0x23b872dd"},
        "setApprovalForAll": {"0xa22cb465"},
        "permit": {
            "0xd505accf",
            "0x8fcbaf0c"
        },
        "safeTransferFrom": {
            "0x42842e0e",
            "0xb88d4fde"
        }
    }

    features = {
        "gas": gas,
        "gas_price_gwei": gas_price_gwei,
        "value_eth": value_eth,
        "nonce": nonce,

        "input_length": input_length,
        "input_word_count": input_word_count,

        "zero_byte_count": zero_byte_count,
        "nonzero_byte_count": nonzero_byte_count,

        "value_is_zero": int(
            value_wei == 0
        ),

        "has_calldata": int(
            input_length > 10
        ),

        "is_contract_call": int(
            bool(to_address) and
            input_length > 10
        ),

        "gas_log": np.log1p(
            max(gas, 0)
        ),

        "gas_price_gwei_log": np.log1p(
            max(gas_price_gwei, 0)
        ),

        "value_eth_log": np.log1p(
            max(value_eth, 0)
        ),

        "nonce_log": np.log1p(
            max(nonce, 0)
        ),

        "input_length_log": np.log1p(
            max(input_length, 0)
        ),

        "input_word_count_log": np.log1p(
            max(input_word_count, 0)
        ),

        "zero_byte_count_log": np.log1p(
            max(zero_byte_count, 0)
        ),

        "nonzero_byte_count_log": np.log1p(
            max(nonzero_byte_count, 0)
        ),
    }

    for name, selectors in SELECTOR_MAP.items():

        features[
            f"selector_{name}"
        ] = int(
            selector in selectors
        )

    known_selector = max(
        features[
            f"selector_{name}"
        ]
        for name in SELECTOR_MAP
    )

    features["known_selector"] = (
        known_selector
    )

    features["unknown_selector"] = (
        1 - known_selector
    )

    return features
```

```python
def aggregate_transaction_risk(scores):

    scores = np.asarray(
        scores,
        dtype=float
    )

    if len(scores) == 0:
        return 0.0

    scores = np.clip(
        scores,
        0,
        100
    )

    top_k = np.sort(
        scores
    )[-5:]

    return float(
        np.clip(
            0.50 * np.max(scores)
            + 0.30 * np.mean(top_k)
            + 0.20 * np.mean(scores),
            0,
            100
        )
    )
```

```python
# ============================================================
# SINGLE EVM TRANSACTION -> MODEL FEATURES
# ============================================================

def transaction_to_model_features(tx):
    import re
    import numpy as np
    import pandas as pd

    gas = float(tx.get("gas") or 0)
    gas_price = float(tx.get("gas_price") or 0)
    value_wei = float(tx.get("value_wei") or 0)
    nonce = float(tx.get("nonce") or 0)

    input_data = tx.get("input") or ""

    if not isinstance(input_data, str):
        input_data = str(input_data)

    input_data = input_data.lower()

    to_address = tx.get("to")

    # --------------------------------------------------------
    # Calldata
    # --------------------------------------------------------

    selector = input_data[:10]

    input_length = len(input_data)

    input_word_count = max(
        (input_length - 10) / 64,
        0
    )

    zero_byte_count = input_data.count("00")

    nonzero_byte_count = len(
        re.findall(
            r"[1-9a-f]",
            input_data
        )
    )

    # --------------------------------------------------------
    # Units
    # --------------------------------------------------------

    value_eth = (
        value_wei / 10**18
    )

    gas_price_gwei = (
        gas_price / 10**9
    )

    # --------------------------------------------------------
    # Features
    # --------------------------------------------------------

    features = {
        "gas": gas,

        "gas_price_gwei": gas_price_gwei,

        "value_eth": value_eth,

        "nonce": nonce,

        "input_length": input_length,

        "input_word_count": input_word_count,

        "zero_byte_count": zero_byte_count,

        "nonzero_byte_count": nonzero_byte_count,

        "value_is_zero": int(
            value_wei == 0
        ),

        "has_calldata": int(
            input_length > 10
        ),

        "is_contract_call": int(
            bool(to_address)
            and input_length > 10
        ),

        "gas_log": np.log1p(
            max(gas, 0)
        ),

        "gas_price_gwei_log": np.log1p(
            max(gas_price_gwei, 0)
        ),

        "value_eth_log": np.log1p(
            max(value_eth, 0)
        ),

        "nonce_log": np.log1p(
            max(nonce, 0)
        ),

        "input_length_log": np.log1p(
            max(input_length, 0)
        ),

        "input_word_count_log": np.log1p(
            max(input_word_count, 0)
        ),

        "zero_byte_count_log": np.log1p(
            max(zero_byte_count, 0)
        ),

        "nonzero_byte_count_log": np.log1p(
            max(nonzero_byte_count, 0)
        ),
    }

    # --------------------------------------------------------
    # Known function selectors
    # --------------------------------------------------------

    SELECTOR_MAP = {
        "approve": {
            "0x095ea7b3"
        },

        "transfer": {
            "0xa9059cbb"
        },

        "transferFrom": {
            "0x23b872dd"
        },

        "setApprovalForAll": {
            "0xa22cb465"
        },

        "permit": {
            "0xd505accf",
            "0x8fcbaf0c"
        },

        "safeTransferFrom": {
            "0x42842e0e",
            "0xb88d4fde"
        }
    }

    for name, selectors in SELECTOR_MAP.items():

        features[
            f"selector_{name}"
        ] = int(
            selector in selectors
        )

    features["known_selector"] = max(
        features[
            f"selector_{name}"
        ]
        for name in SELECTOR_MAP
    )

    features["unknown_selector"] = (
        1 - features["known_selector"]
    )

    return features
```

```python
# ============================================================
# BUILD ml_engine.py
# ============================================================

from pathlib import Path

engine_code = r'''
from __future__ import annotations

from dataclasses import dataclass, asdict
from typing import Any
import json
import os
import re

import joblib
import numpy as np
import pandas as pd


# ============================================================
# CONFIG
# ============================================================

DEFAULT_MODEL_PATH = os.getenv(
    "EVM_RISK_ENGINE_PATH",
    "evm_risk_engine_v1.pkl",
)


# ============================================================
# ENGINE
# ============================================================

class EVMRiskEngine:
    """
    EVM blockchain risk engine.

    Components:
      - Transaction phishing XGBoost
      - Wallet anomaly Isolation Forest
      - Threat-intelligence lookup
      - Graph-risk lookup
      - Transparent risk fusion
    """

    def __init__(self, bundle_path: str = DEFAULT_MODEL_PATH):
        self.bundle_path = bundle_path
        self.bundle = joblib.load(bundle_path)

        self.transaction_model = (
            self.bundle["transaction_model"]
        )

        self.transaction_features = (
            self.bundle["transaction_features"]
        )

        self.wallet_anomaly_model = (
            self.bundle["wallet_anomaly_model"]
        )

        self.wallet_anomaly_scaler = (
            self.bundle["wallet_anomaly_scaler"]
        )

        self.wallet_anomaly_features = (
            self.bundle["wallet_anomaly_features"]
        )

        self.graph_risk_table = (
            self.bundle["graph_risk_table"]
        )

        self.threat_intelligence = (
            self.bundle["threat_intelligence"]
        )

    # ========================================================
    # TRANSACTION FEATURE ADAPTER
    # ========================================================

    @staticmethod
    def transaction_to_features(
        tx: dict[str, Any]
    ) -> dict[str, float]:

        gas = float(tx.get("gas") or 0)
        gas_price = float(
            tx.get("gas_price") or 0
        )
        value_wei = float(
            tx.get("value_wei") or 0
        )
        nonce = float(
            tx.get("nonce") or 0
        )

        input_data = tx.get("input") or ""

        if not isinstance(input_data, str):
            input_data = str(input_data)

        input_data = input_data.lower()

        selector = input_data[:10]

        input_length = len(input_data)

        input_word_count = max(
            (input_length - 10) / 64,
            0
        )

        zero_byte_count = input_data.count("00")

        nonzero_byte_count = len(
            re.findall(
                r"[1-9a-f]",
                input_data
            )
        )

        value_eth = (
            value_wei / 10**18
        )

        gas_price_gwei = (
            gas_price / 10**9
        )

        features = {
            "gas": gas,

            "gas_price_gwei":
                gas_price_gwei,

            "value_eth":
                value_eth,

            "nonce":
                nonce,

            "input_length":
                float(input_length),

            "input_word_count":
                float(input_word_count),

            "zero_byte_count":
                float(zero_byte_count),

            "nonzero_byte_count":
                float(nonzero_byte_count),

            "value_is_zero":
                float(value_wei == 0),

            "has_calldata":
                float(input_length > 10),

            "is_contract_call":
                float(
                    bool(tx.get("to"))
                    and input_length > 10
                ),

            "gas_log":
                float(np.log1p(max(gas, 0))),

            "gas_price_gwei_log":
                float(
                    np.log1p(
                        max(
                            gas_price_gwei,
                            0
                        )
                    )
                ),

            "value_eth_log":
                float(
                    np.log1p(
                        max(
                            value_eth,
                            0
                        )
                    )
                ),

            "nonce_log":
                float(
                    np.log1p(
                        max(nonce, 0)
                    )
                ),

            "input_length_log":
                float(
                    np.log1p(
                        max(input_length, 0)
                    )
                ),

            "input_word_count_log":
                float(
                    np.log1p(
                        max(input_word_count, 0)
                    )
                ),

            "zero_byte_count_log":
                float(
                    np.log1p(
                        max(zero_byte_count, 0)
                    )
                ),

            "nonzero_byte_count_log":
                float(
                    np.log1p(
                        max(nonzero_byte_count, 0)
                    )
                ),
        }

        selector_map = {
            "approve": {
                "0x095ea7b3"
            },

            "transfer": {
                "0xa9059cbb"
            },

            "transferFrom": {
                "0x23b872dd"
            },

            "setApprovalForAll": {
                "0xa22cb465"
            },

            "permit": {
                "0xd505accf",
                "0x8fcbaf0c",
            },

            "safeTransferFrom": {
                "0x42842e0e",
                "0xb88d4fde",
            },
        }

        for name, selectors in (
            selector_map.items()
        ):
            features[
                f"selector_{name}"
            ] = float(
                selector in selectors
            )

        known_selector = max(
            features[
                f"selector_{name}"
            ]
            for name in selector_map
        )

        features["known_selector"] = (
            float(known_selector)
        )

        features["unknown_selector"] = (
            float(1 - known_selector)
        )

        return features

    # ========================================================
    # TRANSACTION SCORE
    # ========================================================

    def score_transaction(
        self,
        tx: dict[str, Any]
    ) -> dict[str, Any]:

        features = self.transaction_to_features(
            tx
        )

        X = pd.DataFrame(
            [features]
        )

        X = X[
            self.transaction_features
        ]

        probability = float(
            self.transaction_model
            .predict_proba(X)[0, 1]
        )

        risk_score = float(
            np.clip(
                probability * 100,
                0,
                100
            )
        )

        return {
            "transaction_risk": round(
                risk_score,
                2
            ),

            "phishing_probability_model": round(
                probability,
                6
            ),

            "tx_hash": tx.get("tx_hash"),

            "from": tx.get("from"),

            "to": tx.get("to"),

            "feature_count": len(
                self.transaction_features
            ),
        }

    # ========================================================
    # WALLET ANOMALY
    # ========================================================

    def score_wallet_anomaly(
        self,
        wallet_features: dict[str, Any]
    ) -> float:

        row = {}

        for feature in (
            self.wallet_anomaly_features
        ):
            row[feature] = float(
                wallet_features.get(
                    feature,
                    0
                ) or 0
            )

        X = pd.DataFrame([row])

        X = (
            X
            .replace(
                [np.inf, -np.inf],
                np.nan
            )
            .fillna(0)
        )

        X_scaled = (
            self.wallet_anomaly_scaler
            .transform(X)
        )

        raw = -self.wallet_anomaly_model \
            .decision_function(X_scaled)[0]

        return float(raw)

    # ========================================================
    # THREAT INTELLIGENCE
    # ========================================================

    def threat_lookup(
        self,
        address: str
    ) -> dict[str, Any]:

        address = (
            str(address)
            .lower()
            .strip()
        )

        result = (
            self.threat_intelligence
            .get(address)
        )

        if result is None:
            return {
                "known_threat": False,
                "sources": [],
                "types": [],
                "threat_score": 0.0,
            }

        return {
            "known_threat": True,
            "sources": result.get(
                "sources",
                []
            ),
            "types": result.get(
                "types",
                []
            ),
            "threat_score": 100.0,
        }

    # ========================================================
    # GRAPH RISK
    # ========================================================

    def graph_lookup(
        self,
        address: str
    ) -> dict[str, Any]:

        address = (
            str(address)
            .lower()
            .strip()
        )

        table = self.graph_risk_table

        matches = table[
            table["address"]
            .astype(str)
            .str.lower()
            .eq(address)
        ]

        if matches.empty:
            return {
                "graph_risk": 0.0,
                "graph_reasons": [],
            }

        row = matches.iloc[0]

        reasons = row.get(
            "graph_reasons",
            []
        )

        if not isinstance(
            reasons,
            list
        ):
            reasons = [str(reasons)]

        return {
            "graph_risk": float(
                row["graph_risk_score"]
            ),

            "graph_reasons": reasons,
        }

    # ========================================================
    # FUSION
    # ========================================================

    @staticmethod
    def fuse(
        transaction_risk: float,
        anomaly_score: float,
        graph_risk: float,
        threat_score: float,
        threat_sources: list[str] | None = None,
        threat_types: list[str] | None = None,
    ) -> dict[str, Any]:

        transaction_risk = float(
            np.clip(
                transaction_risk,
                0,
                100
            )
        )

        anomaly_score = float(
            np.clip(
                anomaly_score,
                0,
                100
            )
        )

        graph_risk = float(
            np.clip(
                graph_risk,
                0,
                100
            )
        )

        threat_score = float(
            np.clip(
                threat_score,
                0,
                100
            )
        )

        score = (
            0.35 * transaction_risk
            + 0.20 * anomaly_score
            + 0.20 * graph_risk
            + 0.25 * threat_score
        )

        # Strong override for independently known threats
        if threat_score >= 100:
            score = max(
                score,
                95.0
            )

        score = float(
            np.clip(
                score,
                0,
                100
            )
        )

        if score >= 80:
            level = "CRITICAL"
        elif score >= 60:
            level = "HIGH"
        elif score >= 35:
            level = "MEDIUM"
        else:
            level = "LOW"

        reasons = []

        if threat_score >= 100:
            reasons.append(
                "Address matched known threat intelligence"
            )

        if transaction_risk >= 80:
            reasons.append(
                "Transaction behavior strongly resembles phishing activity"
            )
        elif transaction_risk >= 60:
            reasons.append(
                "Transaction behavior shows elevated phishing risk"
            )

        if anomaly_score >= 80:
            reasons.append(
                "Wallet behavior is highly anomalous"
            )
        elif anomaly_score >= 60:
            reasons.append(
                "Wallet behavior is unusually structured"
            )

        if graph_risk >= 80:
            reasons.append(
                "Wallet has elevated graph risk"
            )
        elif graph_risk >= 60:
            reasons.append(
                "Wallet shows elevated graph connectivity risk"
            )

        for item in (
            threat_types or []
        ):
            reasons.append(
                f"Threat intelligence: {item}"
            )

        return {
            "risk_score": round(
                score,
                2
            ),

            "risk_level": level,

            "signals": {
                "transaction_risk":
                    round(
                        transaction_risk,
                        2
                    ),

                "anomaly_score":
                    round(
                        anomaly_score,
                        2
                    ),

                "graph_risk":
                    round(
                        graph_risk,
                        2
                    ),

                "threat_intelligence":
                    round(
                        threat_score,
                        2
                    ),
            },

            "weights": {
                "transaction": 0.35,
                "anomaly": 0.20,
                "graph": 0.20,
                "threat_intelligence": 0.25,
            },

            "threat_sources":
                sorted(
                    set(
                        threat_sources or []
                    )
                ),

            "threat_types":
                sorted(
                    set(
                        threat_types or []
                    )
                ),

            "reasons": reasons,
        }

    # ========================================================
    # COMPLETE TRANSACTION ANALYSIS
    # ========================================================

    def analyze_transaction(
        self,
        tx: dict[str, Any]
    ) -> dict[str, Any]:

        transaction = (
            self.score_transaction(tx)
        )

        address = (
            tx.get("from")
            or tx.get("to")
            or ""
        )

        intel = self.threat_lookup(
            address
        )

        graph = self.graph_lookup(
            address
        )

        fused = self.fuse(
            transaction_risk=
                transaction["transaction_risk"],

            anomaly_score=0.0,

            graph_risk=
                graph["graph_risk"],

            threat_score=
                intel["threat_score"],

            threat_sources=
                intel["sources"],

            threat_types=
                intel["types"],
        )

        return {
            "address": address,

            "transaction": transaction,

            "threat_intelligence": intel,

            "graph": graph,

            "risk": fused,
        }


    # ========================================================
    # WALLET ANALYSIS
    # ========================================================

    def analyze_wallet(
        self,
        address: str,
        transactions: list[dict[str, Any]],
        wallet_features: dict[str, Any],
    ) -> dict[str, Any]:

        address = (
            address
            .lower()
            .strip()
        )

        # -----------------------------------------------
        # Transaction scores
        # -----------------------------------------------

        transaction_results = []

        for tx in transactions:

            try:
                transaction_results.append(
                    self.score_transaction(tx)
                )
            except Exception:
                continue

        tx_scores = [
            item["transaction_risk"]
            for item in transaction_results
        ]

        if tx_scores:

            sorted_scores = sorted(
                tx_scores,
                reverse=True
            )

            top5 = sorted_scores[:5]

            transaction_risk = (
                0.50 * sorted_scores[0]
                + 0.30 * np.mean(top5)
                + 0.20 * np.mean(sorted_scores)
            )

        else:
            transaction_risk = 0.0

        # -----------------------------------------------
        # Anomaly
        # -----------------------------------------------

        anomaly_raw = (
            self.score_wallet_anomaly(
                wallet_features
            )
        )

        # Keep same broad 0-100 convention used
        # during notebook scoring.
        # This is intentionally a score, not a probability.

        anomaly_score = float(
            np.clip(
                anomaly_raw * 100,
                0,
                100
            )
        )

        # -----------------------------------------------
        # Threat intelligence
        # -----------------------------------------------

        intel = self.threat_lookup(
            address
        )

        # -----------------------------------------------
        # Graph
        # -----------------------------------------------

        graph = self.graph_lookup(
            address
        )

        # -----------------------------------------------
        # Final fusion
        # -----------------------------------------------

        risk = self.fuse(
            transaction_risk=
                transaction_risk,

            anomaly_score=
                anomaly_score,

            graph_risk=
                graph["graph_risk"],

            threat_score=
                intel["threat_score"],

            threat_sources=
                intel["sources"],

            threat_types=
                intel["types"],
        )

        return {
            "address": address,

            "transaction_count":
                len(transaction_results),

            "transaction_results":
                transaction_results,

            "wallet_anomaly": {
                "score":
                    round(
                        anomaly_score,
                        2
                    )
            },

            "threat_intelligence":
                intel,

            "graph":
                graph,

            "risk":
                risk,
        }
'''

engine_path = Path("/content/evm_models/ml_engine.py")
engine_path.write_text(engine_code)

print("Created:", engine_path)
print("Size:", engine_path.stat().st_size, "bytes")
```

```python
from pathlib import Path

Path("/content/train_crypto_risk_models.py").write_text(r'''
"""
SIH 183 / Crypto Fraud Attribution
==================================

Reusable ML training pipeline.

This file is intentionally independent of FastAPI/backend code.

Current model families:
    1. EVM transaction phishing model
    2. EVM wallet anomaly model
    3. EVM structural/graph model
    4. Bitcoin behavioral model
    5. Threat-intelligence feature preparation
    6. Model artifact export

The pipeline is designed so additional datasets can be plugged in later.
"""

from __future__ import annotations

import json
import os
import warnings
from pathlib import Path
from typing import Any

import joblib
import numpy as np
import pandas as pd

from sklearn.ensemble import (
    IsolationForest,
    RandomForestClassifier,
)
from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    average_precision_score,
    classification_report,
    confusion_matrix,
    precision_recall_curve,
    roc_auc_score,
)
from sklearn.model_selection import (
    GroupShuffleSplit,
    train_test_split,
)
from sklearn.preprocessing import RobustScaler

from xgboost import XGBClassifier


warnings.filterwarnings(
    "ignore",
    category=UserWarning,
)


# ============================================================
# CONFIG
# ============================================================

ROOT = Path("/content")

MODEL_DIR = ROOT / "crypto_risk_models"
MODEL_DIR.mkdir(
    parents=True,
    exist_ok=True,
)

RANDOM_STATE = 42

# EVM transaction dataset produced earlier
EVM_TRANSACTION_FILE = (
    ROOT / "evm_transaction_ml_dataset.parquet"
)

# Graph feature dataset produced earlier.
# The script checks both parquet and csv.
GRAPH_FEATURE_CANDIDATES = [
    ROOT / "graph_feature_dataset.parquet",
    ROOT / "graph_feature_dataset.csv",
]

# Bitcoin feature dataset candidates.
BITCOIN_FEATURE_CANDIDATES = [
    ROOT / "bitcoin_features.parquet",
    ROOT / "bitcoin_features.csv",
]


# ============================================================
# UTILITIES
# ============================================================

def find_existing(
    candidates: list[Path],
) -> Path | None:

    for path in candidates:
        if path.exists():
            return path

    return None


def save_json(
    path: Path,
    obj: dict[str, Any],
) -> None:

    path.write_text(
        json.dumps(
            obj,
            indent=2,
            default=str,
        )
    )


def clean_numeric_frame(
    X: pd.DataFrame,
) -> pd.DataFrame:

    X = X.copy()

    X = X.replace(
        [np.inf, -np.inf],
        np.nan,
    )

    for col in X.columns:

        X[col] = pd.to_numeric(
            X[col],
            errors="coerce",
        )

    return X


def evaluate_classifier(
    name: str,
    model,
    X_test: pd.DataFrame,
    y_test: pd.Series,
) -> dict[str, Any]:

    probability = model.predict_proba(
        X_test
    )[:, 1]

    prediction = (
        probability >= 0.5
    ).astype(int)

    roc = roc_auc_score(
        y_test,
        probability,
    )

    pr = average_precision_score(
        y_test,
        probability,
    )

    print("\n" + "=" * 72)
    print(name)
    print("=" * 72)

    print(
        classification_report(
            y_test,
            prediction,
            target_names=[
                "BENIGN",
                "PHISHING",
            ],
            digits=4,
        )
    )

    print(
        "Confusion matrix:"
    )

    print(
        confusion_matrix(
            y_test,
            prediction,
        )
    )

    print(
        f"ROC-AUC: {roc:.6f}"
    )

    print(
        f"PR-AUC : {pr:.6f}"
    )

    return {
        "roc_auc": float(roc),
        "pr_auc": float(pr),
        "confusion_matrix":
            confusion_matrix(
                y_test,
                prediction,
            ).tolist(),
    }


# ============================================================
# 1. EVM TRANSACTION MODEL
# ============================================================

def train_evm_transaction_model() -> None:

    print(
        "\n"
        + "=" * 72
    )

    print(
        "1. EVM TRANSACTION MODEL"
    )

    print(
        "=" * 72
    )

    if not EVM_TRANSACTION_FILE.exists():

        print(
            f"Dataset not found: "
            f"{EVM_TRANSACTION_FILE}"
        )

        print(
            "Skipping EVM transaction training."
        )

        return

    df = pd.read_parquet(
        EVM_TRANSACTION_FILE
    )

    print(
        "Rows:",
        len(df),
    )

    if "label" not in df.columns:

        raise ValueError(
            "EVM dataset requires 'label'"
        )

    # Exclude identifiers and metadata.
    exclude = {
        "label",
        "tx_hash",
        "address",
        "from",
        "to",
        "category",
    }

    features = [
        c
        for c in df.columns
        if c not in exclude
        and pd.api.types.is_numeric_dtype(
            df[c]
        )
    ]

    X = clean_numeric_frame(
        df[features]
    )

    y = (
        pd.to_numeric(
            df["label"],
            errors="coerce",
        )
        .fillna(0)
        .astype(int)
    )

    X = X.fillna(0)

    # --------------------------------------------------------
    # Group-aware split where possible
    # --------------------------------------------------------

    group_col = None

    for candidate in [
        "from_address",
        "from",
        "address",
    ]:

        if candidate in df.columns:

            group_col = candidate
            break

    if group_col:

        groups = (
            df[group_col]
            .astype(str)
            .str.lower()
        )

        splitter = GroupShuffleSplit(
            n_splits=1,
            test_size=0.25,
            random_state=RANDOM_STATE,
        )

        train_idx, test_idx = next(
            splitter.split(
                X,
                y,
                groups=groups,
            )
        )

        X_train = X.iloc[
            train_idx
        ]

        X_test = X.iloc[
            test_idx
        ]

        y_train = y.iloc[
            train_idx
        ]

        y_test = y.iloc[
            test_idx
        ]

        print(
            "Split: GROUPED"
        )

    else:

        (
            X_train,
            X_test,
            y_train,
            y_test,
        ) = train_test_split(
            X,
            y,
            test_size=0.25,
            random_state=RANDOM_STATE,
            stratify=y,
        )

        print(
            "Split: STRATIFIED"
        )

    neg = int(
        (y_train == 0).sum()
    )

    pos = int(
        (y_train == 1).sum()
    )

    scale_pos_weight = (
        neg / max(pos, 1)
    )

    print(
        "Train:",
        X_train.shape,
    )

    print(
        "Test:",
        X_test.shape,
    )

    print(
        "scale_pos_weight:",
        scale_pos_weight,
    )

    model = XGBClassifier(
        n_estimators=700,
        max_depth=7,
        learning_rate=0.035,
        subsample=0.85,
        colsample_bytree=0.85,
        min_child_weight=2,
        reg_alpha=0.2,
        reg_lambda=2.0,
        objective="binary:logistic",
        eval_metric="aucpr",
        tree_method="hist",
        random_state=RANDOM_STATE,
        n_jobs=-1,
        scale_pos_weight=scale_pos_weight,
    )

    model.fit(
        X_train,
        y_train,
        eval_set=[
            (
                X_test,
                y_test,
            )
        ],
        verbose=False,
    )

    metrics = evaluate_classifier(
        "EVM TRANSACTION — GROUPED",
        model,
        X_test,
        y_test,
    )

    feature_importance = pd.DataFrame({
        "feature": features,
        "importance":
            model.feature_importances_,
    }).sort_values(
        "importance",
        ascending=False,
    )

    print(
        "\nTop features:"
    )

    print(
        feature_importance.head(
            25
        ).to_string(
            index=False
        )
    )

    bundle = {
        "model_type":
            "xgboost_evm_transaction",

        "model":
            model,

        "feature_columns":
            features,

        "metrics":
            metrics,

        "random_state":
            RANDOM_STATE,
    }

    joblib.dump(
        bundle,
        MODEL_DIR
        / "evm_transaction_model_grouped.pkl",
    )

    feature_importance.to_csv(
        MODEL_DIR
        / "evm_transaction_feature_importance.csv",
        index=False,
    )

    save_json(
        MODEL_DIR
        / "evm_transaction_metrics.json",
        metrics,
    )


# ============================================================
# 2. EVM WALLET ANOMALY
# ============================================================

def train_evm_wallet_anomaly() -> None:

    print(
        "\n"
        + "=" * 72
    )

    print(
        "2. EVM WALLET ANOMALY MODEL"
    )

    print(
        "=" * 72
    )

    path = find_existing(
        GRAPH_FEATURE_CANDIDATES
    )

    if path is None:

        print(
            "Graph feature dataset not found."
        )

        print(
            "Skipping wallet anomaly training."
        )

        return

    if path.suffix == ".parquet":
        df = pd.read_parquet(path)
    else:
        df = pd.read_csv(path)

    print(
        "Rows:",
        len(df),
    )

    exclude = {
        "address",
        "node_id",
        "source_graph",
        "known_threat",
        "label",
    }

    features = [
        c
        for c in df.columns
        if c not in exclude
        and pd.api.types.is_numeric_dtype(
            df[c]
        )
    ]

    X = clean_numeric_frame(
        df[features]
    ).fillna(0)

    imputer = SimpleImputer(
        strategy="median"
    )

    X_imp = imputer.fit_transform(
        X
    )

    scaler = RobustScaler()

    X_scaled = scaler.fit_transform(
        X_imp
    )

    model = IsolationForest(
        n_estimators=500,
        max_samples="auto",
        contamination=0.03,
        random_state=RANDOM_STATE,
        n_jobs=-1,
    )

    model.fit(
        X_scaled
    )

    raw_scores = -model.decision_function(
        X_scaled
    )

    p01 = float(
        np.quantile(
            raw_scores,
            0.01,
        )
    )

    p99 = float(
        np.quantile(
            raw_scores,
            0.99,
        )
    )

    calibrated = (
        (raw_scores - p01)
        /
        max(
            p99 - p01,
            1e-12,
        )
        * 100
    )

    calibrated = np.clip(
        calibrated,
        0,
        100,
    )

    result = df[
        [
            "address"
        ]
        if "address" in df.columns
        else []
    ].copy()

    if "address" in result.columns:

        result[
            "anomaly_score"
        ] = calibrated

        result = result.sort_values(
            "anomaly_score",
            ascending=False,
        )

        result.to_csv(
            MODEL_DIR
            / "evm_wallet_anomaly_scores.csv",
            index=False,
        )

    bundle = {
        "model_type":
            "isolation_forest_wallet",

        "model":
            model,

        "imputer":
            imputer,

        "scaler":
            scaler,

        "feature_columns":
            features,

        "calibration": {
            "p01": p01,
            "p99": p99,
        },

        "random_state":
            RANDOM_STATE,
    }

    joblib.dump(
        bundle,
        MODEL_DIR
        / "evm_wallet_anomaly_v4.pkl",
    )

    print(
        "Anomaly model saved."
    )

    print(
        "Score statistics:"
    )

    print(
        pd.Series(
            calibrated
        ).describe()
    )


# ============================================================
# 3. GRAPH / STRUCTURAL MODEL
# ============================================================

def train_graph_structural_model() -> None:

    print(
        "\n"
        + "=" * 72
    )

    print(
        "3. GRAPH STRUCTURAL MODEL"
    )

    print(
        "=" * 72
    )

    path = find_existing(
        GRAPH_FEATURE_CANDIDATES
    )

    if path is None:

        print(
            "Graph feature dataset not found."
        )

        return

    if path.suffix == ".parquet":
        df = pd.read_parquet(path)
    else:
        df = pd.read_csv(path)

    # We only train a supervised graph classifier
    # if actual graph labels are available.

    label_col = None

    for candidate in [
        "label",
        "known_threat",
    ]:

        if candidate in df.columns:

            unique = (
                df[candidate]
                .dropna()
                .unique()
            )

            if len(unique) >= 2:

                label_col = candidate
                break

    if label_col is None:

        print(
            "No valid graph labels found."
        )

        print(
            "Keeping graph model unsupervised for now."
        )

        return

    exclude = {
        label_col,
        "address",
        "node_id",
        "source_graph",
    }

    features = [
        c
        for c in df.columns
        if c not in exclude
        and pd.api.types.is_numeric_dtype(
            df[c]
        )
    ]

    X = clean_numeric_frame(
        df[features]
    ).fillna(0)

    y = (
        pd.to_numeric(
            df[label_col],
            errors="coerce",
        )
        .fillna(0)
        .astype(int)
    )

    # Group by source graph to reduce graph-specific leakage.
    groups = (
        df["source_graph"]
        .astype(str)
        if "source_graph" in df.columns
        else pd.Series(
            np.arange(len(df))
        )
    )

    splitter = GroupShuffleSplit(
        n_splits=1,
        test_size=0.25,
        random_state=RANDOM_STATE,
    )

    train_idx, test_idx = next(
        splitter.split(
            X,
            y,
            groups=groups,
        )
    )

    X_train = X.iloc[
        train_idx
    ]

    X_test = X.iloc[
        test_idx
    ]

    y_train = y.iloc[
        train_idx
    ]

    y_test = y.iloc[
        test_idx
    ]

    neg = int(
        (y_train == 0).sum()
    )

    pos = int(
        (y_train == 1).sum()
    )

    model = XGBClassifier(
        n_estimators=600,
        max_depth=6,
        learning_rate=0.04,
        subsample=0.85,
        colsample_bytree=0.85,
        objective="binary:logistic",
        eval_metric="aucpr",
        tree_method="hist",
        random_state=RANDOM_STATE,
        n_jobs=-1,
        scale_pos_weight=(
            neg / max(pos, 1)
        ),
    )

    model.fit(
        X_train,
        y_train,
        verbose=False,
    )

    metrics = evaluate_classifier(
        "GRAPH STRUCTURAL MODEL",
        model,
        X_test,
        y_test,
    )

    bundle = {
        "model_type":
            "xgboost_graph_structural",

        "model":
            model,

        "feature_columns":
            features,

        "label_column":
            label_col,

        "metrics":
            metrics,
    }

    joblib.dump(
        bundle,
        MODEL_DIR
        / "graph_structural_model.pkl",
    )

    save_json(
        MODEL_DIR
        / "graph_structural_metrics.json",
        metrics,
    )


# ============================================================
# 4. BITCOIN MODEL
# ============================================================

def train_bitcoin_model() -> None:

    print(
        "\n"
        + "=" * 72
    )

    print(
        "4. BITCOIN MODEL"
    )

    print(
        "=" * 72
    )

    path = find_existing(
        BITCOIN_FEATURE_CANDIDATES
    )

    if path is None:

        print(
            "Bitcoin feature dataset not found."
        )

        print(
            "Skipping Bitcoin training."
        )

        return

    if path.suffix == ".parquet":
        df = pd.read_parquet(path)
    else:
        df = pd.read_csv(path)

    if "label" not in df.columns:

        print(
            "Bitcoin dataset has no label."
        )

        return

    exclude = {
        "label",
        "address",
        "tx_hash",
    }

    features = [
        c
        for c in df.columns
        if c not in exclude
        and pd.api.types.is_numeric_dtype(
            df[c]
        )
    ]

    X = clean_numeric_frame(
        df[features]
    ).fillna(0)

    y = (
        pd.to_numeric(
            df["label"],
            errors="coerce",
        )
        .fillna(0)
        .astype(int)
    )

    (
        X_train,
        X_test,
        y_train,
        y_test,
    ) = train_test_split(
        X,
        y,
        test_size=0.25,
        random_state=RANDOM_STATE,
        stratify=y,
    )

    model = XGBClassifier(
        n_estimators=600,
        max_depth=7,
        learning_rate=0.04,
        subsample=0.85,
        colsample_bytree=0.85,
        objective="binary:logistic",
        eval_metric="aucpr",
        tree_method="hist",
        random_state=RANDOM_STATE,
        n_jobs=-1,
    )

    model.fit(
        X_train,
        y_train,
        verbose=False,
    )

    metrics = evaluate_classifier(
        "BITCOIN MODEL",
        model,
        X_test,
        y_test,
    )

    bundle = {
        "model_type":
            "xgboost_bitcoin",

        "model":
            model,

        "feature_columns":
            features,

        "metrics":
            metrics,
    }

    joblib.dump(
        bundle,
        MODEL_DIR
        / "bitcoin_model.pkl",
    )

    save_json(
        MODEL_DIR
        / "bitcoin_metrics.json",
        metrics,
    )


# ============================================================
# 5. RUN EVERYTHING
# ============================================================

def main() -> None:

    print(
        "=" * 72
    )

    print(
        "SIH 183 CRYPTO RISK TRAINING PIPELINE"
    )

    print(
        "=" * 72
    )

    print(
        "Model directory:",
        MODEL_DIR,
    )

    train_evm_transaction_model()

    train_evm_wallet_anomaly()

    train_graph_structural_model()

    train_bitcoin_model()

    print(
        "\n"
        + "=" * 72
    )

    print(
        "TRAINING PIPELINE COMPLETE"
    )

    print(
        "=" * 72
    )

    print(
        "Artifacts:"
    )

    for path in sorted(
        MODEL_DIR.iterdir()
    ):

        print(
            " -",
            path.name,
            f"({path.stat().st_size:,} bytes)"
        )


if __name__ == "__main__":
    main()
''')

print("Created /content/train_crypto_risk_models.py")
```

```python
# ============================================================
# ELLIPTIC++ BITCOIN TRANSACTION MODEL
# TEMPORAL + GROUP-AWARE EVALUATION
# ============================================================

from pathlib import Path
import json
import warnings

import numpy as np
import pandas as pd

from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score,
    average_precision_score,
)

from xgboost import XGBClassifier

warnings.filterwarnings("ignore")

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

ROOT = Path("/content/EllipticPlusPlus")

TX_FEATURES = (
    ROOT
    / "Transactions Dataset"
    / "txs_features.csv"
)

TX_CLASSES = (
    ROOT
    / "Transactions Dataset"
    / "txs_classes.csv"
)

TX_EDGES = (
    ROOT
    / "Transactions Dataset"
    / "txs_edgelist.csv"
)

MODEL_DIR = Path(
    "/content/crypto_risk_models"
)

MODEL_DIR.mkdir(
    parents=True,
    exist_ok=True,
)

# ------------------------------------------------------------
# LOAD
# ------------------------------------------------------------

print("=" * 80)
print("LOADING ELLIPTIC++")
print("=" * 80)

features = pd.read_csv(
    TX_FEATURES
)

classes = pd.read_csv(
    TX_CLASSES
)

edges = pd.read_csv(
    TX_EDGES
)

print(
    "Features:",
    features.shape
)

print(
    "Classes:",
    classes.shape
)

print(
    "Edges:",
    edges.shape
)

# ------------------------------------------------------------
# NORMALIZE CLASS DATA
# ------------------------------------------------------------

classes["class"] = pd.to_numeric(
    classes["class"],
    errors="coerce"
)

classes = classes[
    classes["class"].isin([1, 2, 3])
].copy()

# ------------------------------------------------------------
# MERGE
# ------------------------------------------------------------

df = features.merge(
    classes[
        ["txId", "class"]
    ],
    on="txId",
    how="inner"
)

print(
    "Merged rows:",
    len(df)
)

print(
    "\nClass distribution:"
)

print(
    df["class"].value_counts()
)

# ------------------------------------------------------------
# SUPERVISED DATA
# ------------------------------------------------------------

labeled = df[
    df["class"].isin([1, 2])
].copy()

labeled["label"] = (
    labeled["class"] == 1
).astype(int)

print(
    "\nLabeled rows:",
    len(labeled)
)

print(
    labeled["label"].value_counts()
)

# ------------------------------------------------------------
# TIME FIELD
# ------------------------------------------------------------

TIME_COL = "Time step"

labeled[TIME_COL] = pd.to_numeric(
    labeled[TIME_COL],
    errors="coerce"
)

labeled = labeled.dropna(
    subset=[TIME_COL]
).copy()

labeled[TIME_COL] = (
    labeled[TIME_COL]
    .astype(int)
)

print(
    "\nTime range:",
    labeled[TIME_COL].min(),
    "→",
    labeled[TIME_COL].max()
)

print(
    "Unique time steps:",
    sorted(
        labeled[TIME_COL]
        .unique()
        .tolist()
    )
)

# ------------------------------------------------------------
# FEATURES
# ------------------------------------------------------------

exclude = {
    "txId",
    "class",
    "label",
}

feature_cols = [
    c
    for c in labeled.columns
    if c not in exclude
    and pd.api.types.is_numeric_dtype(
        labeled[c]
    )
]

print(
    "\nNumeric features:",
    len(feature_cols)
)

X_all = labeled[
    feature_cols
].copy()

X_all = X_all.replace(
    [np.inf, -np.inf],
    np.nan
)

y_all = labeled[
    "label"
].astype(int)

# ------------------------------------------------------------
# TEMPORAL SPLIT
# ------------------------------------------------------------

time_steps = sorted(
    labeled[TIME_COL]
    .unique()
)

# Roughly:
# first 70% -> train
# next 15%  -> validation
# last 15%  -> test

n_steps = len(
    time_steps
)

train_end = int(
    n_steps * 0.70
)

valid_end = int(
    n_steps * 0.85
)

train_steps = set(
    time_steps[:train_end]
)

valid_steps = set(
    time_steps[
        train_end:valid_end
    ]
)

test_steps = set(
    time_steps[valid_end:]
)

train_mask = (
    labeled[TIME_COL]
    .isin(train_steps)
)

valid_mask = (
    labeled[TIME_COL]
    .isin(valid_steps)
)

test_mask = (
    labeled[TIME_COL]
    .isin(test_steps)
)

X_train = X_all.loc[
    train_mask
].copy()

X_valid = X_all.loc[
    valid_mask
].copy()

X_test = X_all.loc[
    test_mask
].copy()

y_train = y_all.loc[
    train_mask
]

y_valid = y_all.loc[
    valid_mask
]

y_test = y_all.loc[
    test_mask
]

print(
    "\nTEMPORAL SPLIT"
)

print(
    "Train steps:",
    sorted(train_steps)
)

print(
    "Valid steps:",
    sorted(valid_steps)
)

print(
    "Test steps:",
    sorted(test_steps)
)

print(
    "\nShapes:"
)

print(
    "Train:",
    X_train.shape,
    y_train.value_counts().to_dict()
)

print(
    "Valid:",
    X_valid.shape,
    y_valid.value_counts().to_dict()
)

print(
    "Test:",
    X_test.shape,
    y_test.value_counts().to_dict()
)

# ------------------------------------------------------------
# IMPUTATION
# ------------------------------------------------------------

imputer = SimpleImputer(
    strategy="median"
)

X_train_imp = (
    imputer.fit_transform(
        X_train
    )
)

X_valid_imp = (
    imputer.transform(
        X_valid
    )
)

X_test_imp = (
    imputer.transform(
        X_test
    )
)

# ------------------------------------------------------------
# CLASS WEIGHT
# ------------------------------------------------------------

neg = int(
    (y_train == 0).sum()
)

pos = int(
    (y_train == 1).sum()
)

scale_pos_weight = (
    neg / max(pos, 1)
)

print(
    "\nscale_pos_weight:",
    scale_pos_weight
)

# ------------------------------------------------------------
# MODEL
# ------------------------------------------------------------

model = XGBClassifier(
    n_estimators=800,
    max_depth=7,
    learning_rate=0.035,
    subsample=0.85,
    colsample_bytree=0.85,
    min_child_weight=2,
    reg_alpha=0.2,
    reg_lambda=2.0,

    objective="binary:logistic",

    eval_metric="aucpr",

    tree_method="hist",

    random_state=42,

    n_jobs=-1,

    scale_pos_weight=
        scale_pos_weight,
)

print(
    "\nTraining..."
)

model.fit(
    X_train_imp,
    y_train,
    eval_set=[
        (
            X_valid_imp,
            y_valid
        )
    ],
    verbose=False,
)

print(
    "Training complete."
)

# ------------------------------------------------------------
# EVALUATION
# ------------------------------------------------------------

def evaluate(
    name,
    X,
    y,
):

    prob = model.predict_proba(
        X
    )[:, 1]

    pred = (
        prob >= 0.5
    ).astype(int)

    roc = roc_auc_score(
        y,
        prob
    )

    pr = average_precision_score(
        y,
        prob
    )

    print(
        "\n"
        + "=" * 80
    )

    print(name)

    print(
        "=" * 80
    )

    print(
        classification_report(
            y,
            pred,
            target_names=[
                "LICIT",
                "ILLICIT",
            ],
            digits=4,
        )
    )

    print(
        "Confusion Matrix:"
    )

    print(
        confusion_matrix(
            y,
            pred
        )
    )

    print(
        f"ROC-AUC: {roc:.6f}"
    )

    print(
        f"PR-AUC : {pr:.6f}"
    )

    return {
        "roc_auc":
            float(roc),

        "pr_auc":
            float(pr),

        "confusion_matrix":
            confusion_matrix(
                y,
                pred
            ).tolist(),
    }


valid_metrics = evaluate(
    "ELLIPTIC++ VALIDATION",
    X_valid_imp,
    y_valid,
)

test_metrics = evaluate(
    "ELLIPTIC++ TEMPORAL TEST",
    X_test_imp,
    y_test,
)

# ------------------------------------------------------------
# FEATURE IMPORTANCE
# ------------------------------------------------------------

importance = pd.DataFrame({
    "feature":
        feature_cols,

    "importance":
        model.feature_importances_,
}).sort_values(
    "importance",
    ascending=False
)

print(
    "\nTOP 30 FEATURES"
)

print(
    importance.head(30).to_string(
        index=False
    )
)

importance.to_csv(
    MODEL_DIR
    / "ellipticpp_transaction_feature_importance.csv",
    index=False,
)

# ------------------------------------------------------------
# SAVE BUNDLE
# ------------------------------------------------------------

bundle = {
    "model_type":
        "ellipticpp_bitcoin_transaction_xgboost",

    "model":
        model,

    "imputer":
        imputer,

    "feature_columns":
        feature_cols,

    "label_definition": {
        "0":
            "licit",
        "1":
            "illicit",
    },

    "unknown_class":
        3,

    "metrics": {
        "validation":
            valid_metrics,
        "temporal_test":
            test_metrics,
    },

    "temporal_split": {
        "train":
            sorted(train_steps),
        "validation":
            sorted(valid_steps),
        "test":
            sorted(test_steps),
    },

    "random_state":
        42,
}

import joblib

joblib.dump(
    bundle,
    MODEL_DIR
    / "ellipticpp_bitcoin_transaction_model.pkl",
)

with open(
    MODEL_DIR
    / "ellipticpp_bitcoin_transaction_metrics.json",
    "w",
) as f:

    json.dump(
        {
            "validation":
                valid_metrics,
            "temporal_test":
                test_metrics,
            "temporal_split":
                bundle[
                    "temporal_split"
                ],
        },
        f,
        indent=2,
    )

print(
    "\n"
    + "=" * 80
)

print(
    "ELLIPTIC++ TRANSACTION MODEL SAVED"
)

print(
    MODEL_DIR
    / "ellipticpp_bitcoin_transaction_model.pkl"
)

print(
    "=" * 80
)
```

```python
import json
from pathlib import Path

MODEL_DIR = Path("/content/crypto_risk_models")

def json_safe(obj):
    if isinstance(obj, dict):
        return {
            str(k): json_safe(v)
            for k, v in obj.items()
        }

    if isinstance(obj, (list, tuple)):
        return [
            json_safe(v)
            for v in obj
        ]

    if isinstance(obj, np.integer):
        return int(obj)

    if isinstance(obj, np.floating):
        return float(obj)

    if isinstance(obj, np.ndarray):
        return obj.tolist()

    return obj


safe_metrics = json_safe({
    "validation": valid_metrics,
    "temporal_test": test_metrics,
    "temporal_split": bundle["temporal_split"],
})

with open(
    MODEL_DIR / "ellipticpp_bitcoin_transaction_metrics.json",
    "w"
) as f:
    json.dump(
        safe_metrics,
        f,
        indent=2,
    )

print("Metrics saved successfully.")
```

```python
# ============================================================
# ELLIPTIC++ BITCOIN INTERPRETABLE BEHAVIOR MODEL
# ============================================================

from pathlib import Path
import json

import numpy as np
import pandas as pd

from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score,
    average_precision_score,
)
from xgboost import XGBClassifier


ROOT = Path("/content/EllipticPlusPlus")

TX_FEATURES = (
    ROOT
    / "Transactions Dataset"
    / "txs_features.csv"
)

TX_CLASSES = (
    ROOT
    / "Transactions Dataset"
    / "txs_classes.csv"
)

MODEL_DIR = Path(
    "/content/crypto_risk_models"
)

MODEL_DIR.mkdir(
    parents=True,
    exist_ok=True,
)

# ------------------------------------------------------------
# LOAD
# ------------------------------------------------------------

features = pd.read_csv(
    TX_FEATURES
)

classes = pd.read_csv(
    TX_CLASSES
)

df = features.merge(
    classes[
        ["txId", "class"]
    ],
    on="txId",
    how="inner",
)

# Only known licit / illicit transactions
df = df[
    df["class"].isin([1, 2])
].copy()

df["label"] = (
    df["class"] == 1
).astype(int)

df["Time step"] = pd.to_numeric(
    df["Time step"],
    errors="coerce",
)

# ------------------------------------------------------------
# EXPLICIT BEHAVIOR FEATURES
# ------------------------------------------------------------

behavior_features = [
    "Time step",

    "in_txs_degree",
    "out_txs_degree",

    "total_BTC",
    "fees",
    "size",

    "num_input_addresses",
    "num_output_addresses",

    "in_BTC_min",
    "in_BTC_max",
    "in_BTC_mean",
    "in_BTC_median",
    "in_BTC_total",

    "out_BTC_min",
    "out_BTC_max",
    "out_BTC_mean",
    "out_BTC_median",
    "out_BTC_total",
]

behavior_features = [
    c
    for c in behavior_features
    if c in df.columns
]

print(
    "Behavior features:",
    len(behavior_features),
)

print(
    behavior_features
)

X = df[
    behavior_features
].copy()

X = X.replace(
    [np.inf, -np.inf],
    np.nan,
)

y = df[
    "label"
].astype(int)

# ------------------------------------------------------------
# TEMPORAL SPLIT
# ------------------------------------------------------------

time_steps = sorted(
    df["Time step"]
    .dropna()
    .astype(int)
    .unique()
)

n = len(time_steps)

train_end = int(
    n * 0.70
)

valid_end = int(
    n * 0.85
)

train_steps = set(
    time_steps[:train_end]
)

valid_steps = set(
    time_steps[
        train_end:valid_end
    ]
)

test_steps = set(
    time_steps[
        valid_end:
    ]
)

train_mask = (
    df["Time step"]
    .isin(train_steps)
)

valid_mask = (
    df["Time step"]
    .isin(valid_steps)
)

test_mask = (
    df["Time step"]
    .isin(test_steps)
)

X_train = X.loc[
    train_mask
]

X_valid = X.loc[
    valid_mask
]

X_test = X.loc[
    test_mask
]

y_train = y.loc[
    train_mask
]

y_valid = y.loc[
    valid_mask
]

y_test = y.loc[
    test_mask
]

print("\nSPLITS")

print(
    "Train:",
    X_train.shape,
)

print(
    "Valid:",
    X_valid.shape,
)

print(
    "Test:",
    X_test.shape,
)

print(
    "Train labels:",
    y_train.value_counts().to_dict()
)

print(
    "Valid labels:",
    y_valid.value_counts().to_dict()
)

print(
    "Test labels:",
    y_test.value_counts().to_dict()
)

# ------------------------------------------------------------
# IMPUTE
# ------------------------------------------------------------

imputer = SimpleImputer(
    strategy="median"
)

X_train_imp = (
    imputer.fit_transform(
        X_train
    )
)

X_valid_imp = (
    imputer.transform(
        X_valid
    )
)

X_test_imp = (
    imputer.transform(
        X_test
    )
)

# ------------------------------------------------------------
# CLASS WEIGHT
# ------------------------------------------------------------

neg = int(
    (y_train == 0).sum()
)

pos = int(
    (y_train == 1).sum()
)

scale_pos_weight = (
    neg / max(pos, 1)
)

print(
    "scale_pos_weight:",
    scale_pos_weight
)

# ------------------------------------------------------------
# TRAIN
# ------------------------------------------------------------

model = XGBClassifier(
    n_estimators=700,
    max_depth=5,
    learning_rate=0.035,
    subsample=0.85,
    colsample_bytree=0.90,

    min_child_weight=3,

    reg_alpha=0.3,
    reg_lambda=2.0,

    objective="binary:logistic",

    eval_metric="aucpr",

    tree_method="hist",

    random_state=42,

    n_jobs=-1,

    scale_pos_weight=
        scale_pos_weight,
)

print(
    "\nTraining interpretable model..."
)

model.fit(
    X_train_imp,
    y_train,
    eval_set=[
        (
            X_valid_imp,
            y_valid
        )
    ],
    verbose=False,
)

print(
    "Training complete."
)

# ------------------------------------------------------------
# EVALUATE
# ------------------------------------------------------------

def evaluate(
    name,
    X_eval,
    y_eval,
):

    probability = (
        model.predict_proba(
            X_eval
        )[:, 1]
    )

    prediction = (
        probability >= 0.5
    ).astype(int)

    roc = roc_auc_score(
        y_eval,
        probability,
    )

    pr = average_precision_score(
        y_eval,
        probability,
    )

    print(
        "\n"
        + "=" * 80
    )

    print(name)

    print(
        "=" * 80
    )

    print(
        classification_report(
            y_eval,
            prediction,
            target_names=[
                "LICIT",
                "ILLICIT",
            ],
            digits=4,
        )
    )

    print(
        "Confusion Matrix:"
    )

    print(
        confusion_matrix(
            y_eval,
            prediction,
        )
    )

    print(
        f"ROC-AUC: {roc:.6f}"
    )

    print(
        f"PR-AUC : {pr:.6f}"
    )

    return {
        "roc_auc": float(roc),
        "pr_auc": float(pr),
        "confusion_matrix":
            confusion_matrix(
                y_eval,
                prediction,
            ).tolist(),
    }


valid_metrics = evaluate(
    "INTERPRETABLE MODEL — VALIDATION",
    X_valid_imp,
    y_valid,
)

test_metrics = evaluate(
    "INTERPRETABLE MODEL — TEMPORAL TEST",
    X_test_imp,
    y_test,
)

# ------------------------------------------------------------
# IMPORTANCE
# ------------------------------------------------------------

importance = pd.DataFrame({
    "feature":
        behavior_features,

    "importance":
        model.feature_importances_,
}).sort_values(
    "importance",
    ascending=False,
)

print(
    "\nFEATURE IMPORTANCE"
)

print(
    importance.to_string(
        index=False
    )
)

importance.to_csv(
    MODEL_DIR
    / "ellipticpp_bitcoin_behavior_importance.csv",
    index=False,
)

# ------------------------------------------------------------
# SAVE
# ------------------------------------------------------------

bundle = {
    "model_type":
        "ellipticpp_bitcoin_behavior_xgboost",

    "model":
        model,

    "imputer":
        imputer,

    "feature_columns":
        behavior_features,

    "metrics": {
        "validation":
            valid_metrics,

        "temporal_test":
            test_metrics,
    },

    "label_definition": {
        "0":
            "licit",

        "1":
            "illicit",
    },

    "unknown_class":
        3,
}

import joblib

joblib.dump(
    bundle,
    MODEL_DIR
    / "ellipticpp_bitcoin_behavior_model.pkl",
)

safe = {
    "validation":
        valid_metrics,

    "temporal_test":
        test_metrics,
}

with open(
    MODEL_DIR
    / "ellipticpp_bitcoin_behavior_metrics.json",
    "w",
) as f:

    json.dump(
        safe,
        f,
        indent=2,
    )

print(
    "\nMODEL SAVED:"
)

print(
    MODEL_DIR
    / "ellipticpp_bitcoin_behavior_model.pkl"
)
```
